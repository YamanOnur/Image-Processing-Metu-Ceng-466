# Chapter 7: Multiresolution Image Analysis

In Chapter 6, we studied some useful transforms, namely, **Fourier, Cosine and Hadamard transforms**. These techniques transform an image from the space domain to some frequency domains spanned by a set of periodic basis functions, with different periods. When we represent an image in the frequency domain, we can analyze it in terms of a set of oscillations in varying amplitudes and frequencies. We can, also, design precise filtering algorithms to enhance  images. However, there are some problems with the classical transform domain techniques, concerning image processing systems. Few of them are outlined below:

-  When the basis functions are periodic, such as complex exponential functions, cosine and Hadamard functions, they are defined in infinite duration. However, images have always finite domains and ranges. Truncating them into a finite interval, introduces some error to the transform and inverse transform equations, which may  result in undesired effects, such as [Gibbs phenomenon](https://en.wikipedia.org/wiki/Gibbs_phenomenon).

-  Recall that sine and cosine functions are continuous functions. Discontinuities of a function cannot be represented in terms of linear combination of continuous basis functions. Unfortunately, images frequently contain sharp corners, lines, boundaries and edges. When we decompose these discontinuities into its frequency components as a linear combination of continuous functions, we introduce an error to the representation.


-  Transform domain techniques loose the spatial information about the image. Our human visual system perceives the physical world in the space domain. Analysis in the frequency domain lacks the visual information.

-  In most applications, appearance of an object or a scene depends on the spatial or radiometric resolution of the image. Small objects disappear in low resolution images, while large objects lose their integrity in high resolution As an example, consider the Google Earth images of Istanbul. If our goal is to discriminate among the rural areas, forests and sea, the resolution of the image in Figure 1.a is sufficient. As we increase resolution (Figure 1.b), we start to lose the integrity of land and sea due to the irrelevant details. On the other hand, if we need to detect the bridge, a higher resolution image is necessary. In order to detect the buildings and the cars, even higher resolution images are needed (Figure 1.c).

Coarse-to-fine representation of images in multiple resolutions is very useful in many image processing applications. Because; each resolution contains a different type of context. While the  high resolution images contains information about the details and small objects, the low resolutions images contains information about the basic shapes and relatively large objects.

  

Developing image processing algorithms in a range of different resolutions is called **multiresolution analysis**. There are three major approaches for multi-resolution analysis:

1. Wavelets transforms

2. Image pyramids and trees

3. Sub band coding

  

In this Chapter, we shall briefly study the above  methods for multiresolution image processing.

# 1. Wavelet Transforms

In this section, we shall introduce two dimensional Discrete Wavelet Transforms (DWT) to decompose an image into multiple resolutions.

**Goal:** Define a set of discrete Wavelet basis functions, which decomposes a digital image into multiple resolutions.

In order to achieve the above goal, we need to define a set of basis functions at multiple scale, such that, we can represent a digital image in terms of the linear combination of the multiple scale basis functions, where the weights (coordinates) of each basis function shows the amount of that particular scale in the image.

Since our images are digital, we can convert the above task to a **matrix similarity transformation** problem, as in the Fourier, Cosine and Hadamard transforms.

Recall that matrix similarity transforms are linear maps, which transform a digital image represented as a matrix, to another space spanned by a predefined set of basis functions. The basis functions characterize the space and representation of an image in this space.

Mathematically speaking, our goal is to design a matrix, $Q$, such that the transformed matrix,

$$F_W=QfQ^{-1} \Leftrightarrow f=Q^{-1}F_WQ,$$

represent the image $f$, in a space spanned by a set of basis vectors with different scales. In other words, the transformation decomposes the image into multiple resolutions. Furthermore, in order to assure invertibility, $Q$ should be an orthogonal matrix, so that $Q^{-1}= Q^{T}$

**Motivating Question:** How do we define $Q$, such that $F_W$ represents the transformed image in multiple resolutions.

The answer behind this question lies in a matrix similarity transform, called Wavelet transforms. The following loose mathematical derivations explains the basic idea for finding the Wavelet transform matrix, $Q$. The derivations are rather incomplete and imprecise. However, it helps us to gain insight about the multiresolution representation  of Wavelet transforms, as well as, their relationship to other transform domains.

**A historical Note:** A special form of continuous  Wavelet transform is first introduced by [Haar](https://en.wikipedia.org/wiki/Alfr%C3%A9d_Haar) in 1909. Later, [Mallat](https://en.wikipedia.org/wiki/St%C3%A9phane_Mallat) generalized the Wavelet transform to a large variety of wavelet functions, in 1987 to represent continuous functions in a space spanned by wavelets. In 1988-89, wavelet transforms were expanded to the discrete functions as **matrix similarity transforms**, by [Daubechies](https://en.wikipedia.org/wiki/Ingrid_Daubechies) and [Mallat](https://en.wikipedia.org/wiki/St%C3%A9phane_Mallat).

For simplicity, let us start by explaining the concept of wavelets for one dimensional continuous functions, then, we  extend the definitions to two dimensional discrete functions.

Recall that, the Fourier space is spanned by the infinitely many basis of complex exponentials, for continuous functions. The complex exponentials are periodic functions in a complex plane which consists of **waves of sines and cosines**, in infinitely many frequencies, some of which are shown in Figure 2. The domain of these periodic functions lie in $x \in [-\infty, \infty].$

## Basic Wavelet Function

  

Consider a continuous  function and its Fourier transform, given below:,

$$\phi(x) \leftrightarrow \Phi(u).$$

  

A function, $\phi(x)$, is called  **admissible**, it it satisfies the following inequality;

  

$$\displaystyle\int_{-\infty}^{\infty} \frac{|\Phi(u)|^2}{|u|}du \lt \infty.$$

A **basic wavelet** can be defined as any finite duration function, $\phi(x)$ , which is real and admissible.

  

There are many basic wavelets in the literature, which are used to generate **basis functions** of finite durations, suggested by many researchers, some of which are shown in figure 3. Basis functions depend on the selection of the basic wavelet function. Therefore, there are many wavelet spaces spanned by different basis functions, generated by different basic functions.

## Generation of Basis Functions from a Basic Wavelet Function

  

A set of wavelet basis functions, $\phi_{s,l}(x)$, can be generated by scaling and translating a predefined basic wavelet function,$\phi(x)$, as follows;

  

$$\phi_{s,l} (x) = \frac{1}{\sqrt{s}}\phi(\frac{x-l}{s}).$$

  

In the above formulation, there are two parameters, $l$ and $s$, to generate wavelet basis functions. Parameter, $l$, translates the basic wavelet, while the parameter, $s$, scales it. For each pair of $(l,s)$ parameter, we generate a basis wavelet function. Therefore, it is possible to generate infinitely many basis functions by selecting $(l,s)$ pairs, for continuous functions.

  

**Note:** The basis functions generated by a basic function span a function space, where each basis represents a different scale and translation of the basic wavelet function. Therefore, the wavelet basis functions enable us to represent a function in multiple scales.

  

Consider the basic wavelet function of Figure 6, suggested by Morlet. By scaling and shifting the Morlet function, it is possible to generate basis functions, which span the [Morlet wavelet space](https://en.wikipedia.org/wiki/Morlet_wavelet).

##  One Dimensional Continuous Wavelet Transforms

  

For the time being, let us  consider the multi-scale representation of one-dimensional continuous functions, for simplicity.

In order to span a function space, the wavelet basis functions, $\phi_{s_i,l_i}(x)$, generated by each pair of parameters $(s_i,l_i)$ defined in the previous section, are to be orthogonal. In other words, their inner product is to be equal to Kronecker delta. Formally speaking, the inner product of $\phi_{s_i,l_i}(x)$ and $\phi_{s_j,l_j}(x)$ is,

  
  

$$<\phi_{s_i,l_i}(x), \phi_{s_j,l_j}(x)> = \int \phi_{s_i,l_i}(x)\phi_{s_j,l_j}(x)dx=\delta(i,j),$$

  

where the [Kronecker delta](https://en.wikipedia.org/wiki/Kronecker_delta) is defined as,

  

$$\delta(i,j)= \begin{cases}1& \text{for } i=j\\ 0 & \text{otherwise}\end{cases}$$

  





Recall one-dimensional Fourier transform of continuous function $f(x)$ and its inverse transform, given in Chapter 6:

$$F(u) \triangleq \int _{-\infty}^{\infty} { f(x) \exp^ {-2\pi jux} dx} \Leftrightarrow f(x) = \int_{-\infty}^{\infty} { F(u) \exp^ {2\pi jux} du},$$

where $\exp^ {-2\pi jux}$ are the orthogonal basis functions in frequency $u$.

Now, we may replace the complex exponential basis functions by the real basis wavelet functions, generated by the basic wavelet, $\phi(x)$.

**Note:** It can be shown that the wavelet basis functions generated by a basic wavelet span a wavelet space.

**Definition:** In the wavelet space, characterized by the basic function $\Phi(x)$, **one-dimensional continuous wavelet transform** of $f(x)$, is defined as follows;

$$F_\phi(s,l) = \int_{-\infty}^{\infty}f(x)\phi_{s,l}(x)dx,$$

where

$$\phi_{s,l}(x) = \frac{1}{\sqrt{s}}\phi(\frac{x-l}{s}),$$

for all real numbers, $s>0,l>0$.

The function $f(x)$ can be uniquely recovered from the inverse wavelet transform, defined below:

$$f(x) = \frac{1}{C_\Phi}\int_{s=0}^{\infty}\int_{l=-\infty}^{\infty} F_\Phi(s,l)\frac{\Phi_{sl}(x)}{s^2}ds dl.$$

In the above equation,

$$C_\phi = \int_{-\infty}^{\infty} \frac{|\mathcal{F}(\Phi(u))|^2}{|u|}du,$$

is a normalization factor and $\mathcal{F}(\Phi(u))$, is the Fourier transform of the basic wavelet, $\phi(x)$. Since the basic wavelet function is admissible, $C_\phi$ is finite. Thus, we can uniquely obtain the inverse wavelet transform.

Comparison of Fourier transform and Wavelet transform equations shows that they transform a function $f(x)$ into two different function spaces. While Fourier transform is a representation of $f(x)$ in a function space, spanned by complex exponentials, the wavelet transform is a representation of the same function in a function space, spanned by scaled and translated wavelet basis. While the Fourier transform represents a function in terms of its frequencies, the wavelet transform represents a function in terms of different scales, in other words, in different resolutions.

**Note ** Since there are two parameters in the wavelet basis functions, the inverse transform requires double integration.



## Two Dimensional Continuous Wavelet Transform

The above one dimensional continuous wavelet transforms can be extended into the transform of two dimensional continuous functions, $f(x,y)$. This time, we need to define a two dimensional basic wavelet function, $\Phi(x,y)$. Since the basic wavelet is two dimensional, the wavelet basis functions should be generated by translating the basis wavelet function in $x$ and $y$ directions as follows:

$$\phi_{s,l_x,l_y} (x,y) = \frac{1}{\lvert s\rvert}\phi(\frac{x-l_x}{s},\frac{y-l_y}{s}).$$

Now, we can extend the one dimensional wavelet transformation into two dimensional functions, as follows:

**Definition: Wavelet transform of a two dimensional continuous function**, $f(x,y),$ with respect to the basic wavelet, $\phi(x,y),$ is defined as follows;

$$F_\phi(s,l_x,l_y) = \int_{-\infty}^{\infty} \int_{-\infty}^{\infty} f(x,y) \phi_{s,l_x,l_y}(x,y) dx dy.$$

The function, $f(x,y)$, can be uniquely obtained from its wavelet transform by taking the inverse, as follows;

$$f(x,y) = \frac{1}{c_\phi} \int_{s=0}^{\infty} \int_{l_x = -\infty}^{\infty} \int_{-\infty}^{\infty} F_\phi (s,l_x,l_y) \phi_{s,l_x,l_y}(x,y)dl_xdl_y\frac{d_s}{s^3}.$$

In the above equation,

$$C_\phi = \int_{-\infty}^{\infty} \int_{-\infty}^{\infty} \frac{|(\Phi(u,v))|^2}{|uv|}du dv,$$

is a normalization factor and $\Phi(u,v)$, is the Fourier transform of the basic wavelet, $\phi(x,y)$.

The derivations of the above pair of equations are beyond the scope of this course. The interested reader can refer to [https://link.springer.com/chapter/10.1007/978-1-4612-1768-8_2](https://link.springer.com/chapter/10.1007/978-1-4612-1768-8_2) .

The above pair of equations for wavelet transform and its inverse look very complicated. Since we need translation in two dimensions, the inverse transform requires to take triple integrals. However, knowing that wavelet transform is a representation of a function $f(x,y)$, in a new space called **wavelet space** simplifies our look at the analysis and synthesis equations of wavelet transform.

As in the previous transforms, the function, f(x,y), corresponds to the coordinate of the transform function, $F_\phi(s,l_x,l_y)$, in the wavelet space, spanned by a set of multiresolution basis functions.

**Note**: Recall that our images are digital. Therefore, the above definition of the two dimensional wavelet transform analysis and synthesis equations are not usable for discrete functions. They are provided just for completeness. We need to define discrete wavelet transforms, represented in terms of matrix similarity transforms.

Before representing a two dimensional discrete wavelet transform and its representation as a matrix similarity transform, let us briefly review the wavelet series, in the following section.



## Wavelet Series for One Dimensional Continuous Functions

  

Wavelet series represent a one dimensional function, $f(x)$ as a linear combination of wavelet basis functions.

  

Recall that, in order to generate  a set of basis wavelet functions, we need to define

-  a basic wavelet function and

-  a scaling and translation protocol to generate the basis functions.

  

In the above sections, we mentioned that any admissible finite duration real function can serve as a basic wavelet. Then, the remaining question isto define a scaling and translation protocol.


**Motivating Question:** How do we define a scaling and translation protocol?

A very popular method for scaling and translation of a basic wavelet is **binary scaling and dyadic translation**. This approach generates a set of orthogonal **dyadic wavelet basis functions**, as explained below.

  


### Dyadic Wavelets as a Set of Basis Functions

  

Dyadic wavelet basis functions are generated by binary scaling and dyadic translation of the basic wavelet, $\phi(x)$.

  

**Binary scaling** iteratively scales the basic wavelet by a factor of 2.

**Dyadic translation** shifts the basic wavelet function iteratively by integer multiples of binary scale factor, $k =2^j\times j$. Mathematically speaking, the dyadic basis functions are generated from a basic function as follows;

  
  

$$\phi_{jk}(x) = 2^{\frac{j}{2}} \phi(2^j x-k),$$

where $j$ and $k$ are integer valued parameters of the wavelet basis function, generated by the relation, $k =2^j\times j$, for $-\infty<(j,k)<\infty$ are integer values.

.

### Wavelet Series

  

Wavelet series expansion of $f(x)$ is defined by using dyadic basis functions, $\phi_{j,k}(x)$ , as follows;

  

$$ f(x)=\sum_j\sum_k c_{jk} \phi_{jk}(x) .$$

  

Wavelet series coefficients can be uniquely obtained from the inner product between the function $f(x)$ and the wavelet basis functions, as follows:

  

$$ c_{jk} = <f(x),\phi_{j,k}(x)> = 2^{j/2} \int_{-\infty}^{\infty}f(x)\phi(2^{j/2}x-k)dx .$$

  

Wavelet series coefficients, $c_{j,k}$ represent the coordinate of the function $f(x)$, in a wavelet space spanned by multiresolution dyadic basis functions, $\Phi_{j,k}(x)$. Therefore, each coordinate $c_{j,k}$, can be considered as the amount of the corresponding basis function, $,\phi_{j,k}(x)$, at scale, $j$, in the function, $f(x)$.

  

Recall that the translation, $k$ is dyadically obtained from the scale, $j$. Note that the wavelet series coefficients, $c_{jk}$ are discrete for a continuous function, $f(x)$, is continuous.



### Compact Dyadic Wavelets

In the above representation there are two indexes, $j,k$, for computing $c_{j,k}$, We can restrict the basic wavelet and the function, $f(x)$, further to define the wavelet coefficients depending on one index, $n$, as $c_n$, only. The set of the basis functions, which depend on just one index, are called **compact dyadic wavelets**.

Mathematically speaking, let $f(x)=0$ and $\phi(x)=0$ for $x\notin[0,1]$. Furthermore, define an integer index, which relates the scaling and translating indexes, $j$ and $k$, as follows;

$n=2^j+k-1,$ for $j=0,\ldots, k$ and $k=0, \ldots,(2^j-1).$

The newly defined index $n$, reduces two indices $k$ and $j$ to one index.

**Note**: For $\forall n$, the index $j$ is the largest integer such that, $2^j<n$ and $k=n-2^j+1$.

Therefore, index $n$ enables us to represents all possible values of $j$ and $k$, in a compact way, as follows;

| | | | | | |
|---|---|---|---|---|---|
| n | 0 | 1 | 2 | 3 | ...|
| j | 0 | 0 | 1 | 1 | ...|
| k | 0 | 1 | 1 | 2 | ...|

Then, we can define compact dyadic wavelet with only one index, $n$, as follows;

$$\phi_{j,k}(x) = \phi_{n}(x) = 2^{j/2}\phi(2^{j/2}x-k).$$

The wavelet series representation of a function, $f(x)$, becomes,

$$f(x) = \sum_{n=0}^{\infty}c_n\phi_n(x),$$

where the wavelet series coefficient can be obtained as follows;

$$c_n=2^{j/2}\int f(x)\phi_n(x)dx.$$

**Motivating Question:** How do we read wavelet series representation and series coefficient equations?

Wavelet series basically represents a function, $f(x)$, in terms of linear combination of basis wavelet functions. Recall that basis functions are obtained by scaling and translating a basic function. Therefore, it decomposes a function, $f(x)$, into a set of wavelet bases at **multiple scales**, which enables us **multiresolution representation of images**.

The coordinate, $c_n$, of each basis function measures the amount of that basis function at its scale. Therefore, wavelet series coefficients can be considered as a measure of a certain resolution, embedded in a function, $f(x)$.



### Continuous Haar Functions: Compact Dyadic Wavelets Basis Functions

  

Recall that wavelet transform is defined with respect to a predefined real and admissible wavelet basic function. There are dozens of different wavelet basic functions, which can be used to generate wavelet basis functions, as shown in Figure 4. Thus, there are many different wavelet transforms defined in different wavelet spaces, spanned by a set of different basis functions.

  

The oldest and the simplest wavelet transform is called the **Haar transform**, suggested by a Hungarian Mathematician, [[Alfréd Haar](https://en.wikipedia.org/wiki/Alfr%C3%A9d_Haar)](https://en.wikipedia.org/wiki/Alfr%C3%A9d_Haar). Haar transforms employ compact dyadic wavelet basis functions, called Haar functions, to decompose a function, $f(x)$ into its coarse-to-fine components, using scaled and translated rectangular functions.

  

**Definition: Haar functions,** $h_n(x),\quad$ for $n= 0,.....,(N-1)$ are defined on the interval $0\leq x \leq 1$, as follows;

  

$$h_0(x)=h_{00}(x) = \frac{1}{\sqrt{N}}$$

  

$$h_n(x) = h_{jk}(x) =\begin{cases} \frac{1}{\sqrt{N}}2^{j/2} & \frac{k-1}{2^j}\leq x < \frac{k-0.5}{2^j} \\ -\frac{1}{\sqrt{N}}2^{j/2} & \frac{k-0.5}{2^j}\leq x < \frac{k}{2^j}\\ 0 & \text{otherwise, } x\in [0,1]\end{cases}.$$

  

**Motivating Question:** What does the above Haar function look like for $n= 0, 1, 2, ....?$

The coarsest Haar Function, $h_0(x)=h_{00}(x) = \frac{1}{\sqrt{N}}$ has just a constant value, for $n=0$.

For $n=1$, we have $k=0$ and $j=1$ and the Haar function becomes;

$$h_1(x) = h_{01}(x) =\begin{cases} \frac{1}{\sqrt{N}}& 0\leq x < 0.5 \\ -\frac{1}{\sqrt{N}} & 0.5\leq x < 1\\ 0 & \text{otherwise, } x\in [0,1]\end{cases}.$$

As $n$ increases, the Haar function is scaled and translated, by **dyadic rules**, mentioned in the previous section, as shown in Figure 8. Therefore, Haar space can be considered as a space, spanned by a set of basis functions at multiple-resolution, due to the different scales of the each Haar basis function.

### Two-Dimensional Discrete Haar Transform

Two-dimensional discrete Haar transform is a two-dimensional discrete compact dyadic wavelet transform.

Recall that, in order to map an image from the space domain to some other domains, we need to define an orthogonal matrix so that we can apply matrix similarity transform for the mapping. This time, our goal is to define $N\times N$ Haar transform matrix, $H_{N}$, so that by using matrix similarity transforms, we can map a two dimensional digital image into a space spanned by a set of multi-scale basis functions.

Mathematically speaking, our goal is to find the Haar matrix $H_N$, such that,

$$F_H=H_{N}fH_{N}^T \leftrightarrow f=H_{N}^TF_HH_{N},$$

maps an $N\times N$ image matrix, $f$, into a space spanned by the multi-scale Haar basis functions.

This is like the magic of pulling a rabbit out of the hat :)

Let's use the magic of Mathematics:

A very nice property of compact dyadic wavelets, such as Haar functions, is that we can define the transform matrix, $H_N$ by directly digitizing the variable $x$ of $h_n(x).$ Then, we define the Haar matrix by the digitized one dimensional Haar functions, in multiple scale. Thus, each row of the Haar matrix represents the digitized one dimensional Haar function at a certain scale.

Once we define the Haar matrix, we can transform an $N\times N$ image, $f$, into a multiscale Haar space, by matrix similarity transform, 

$$F_H=H_NfH_N^T \leftrightarrow f=H_N^TF_HH_N.$$

Formally speaking, the first step is to digitize the one dimensional Haar functions, $h_n(x)$ by sampling $x$ at the following discrete set of points,

$$x=\frac{0}{N}, \frac{1}{N}, \ldots, \frac{N-1}{N}.$$ 

Then, for each scale index $n$, we obtain a vector of $N$ dimension, as follows;

$$h_n = [h_n(0), h_n(1/N), ……., h_n(N-1/N)].$$

Since Haar basis functions, $h_n(x)$, for $n=1,...., (N-1)$, are compact dyadic wavelets, scaling and translation indices $(j,k)$, are compactly represented by a single index , $n$.

**Definition:** Haar matrix, $H_N$ of dimension $N\times N$, is defined as;

$$H_N = \begin{bmatrix}h_0\\ h_1 \\ h_2\\
.\\.\\.\\h_{N-1}\end{bmatrix},$$

  
  

where the $nth$ row of the Haar matrix consists of the entries of the vector, $h_n$. The above  Haar matrix is orthogonal but not symmetric. Thus, $H^{-1}  = H^T.$ We  normalize $x\in[0,1]$, for the  Haar basis functions and image function, $f(x,y$), where the size of the image is $N\times N=2^n\times 2^n$.

## Two Dimensional Extension of Wavelet Transform for Discrete Functions

  

The above derivations for two dimensional Haar transforms, can be generalized for any admissible wavelet basic function, $\phi(x,y),$ where we can generate a set of basis functions, $\phi_{s,l}(x,y),$ at different scale and translation, which spans a wavelet space.

  

There is an elegant mathematics for deriving generic discrete wavelet transforms, which is valid for a wide range of basic wavelet functions. However, at this point, we shall stop and refer the reader to [this link]([https://en.wikipedia.org/wiki/Discrete_wavelet_transform](https://en.wikipedia.org/wiki/Discrete_wavelet_transform)).

  

Generally speaking, discrete wavelet transform of an $N\times N$ matrix, $f$, over a set of basis functions, generated by basic wavelet $\phi$, is defined as follows;

  

$$F_W= W_{\phi} f W_{\phi}^T \leftrightarrow f=W_{\phi}^TF_W W_{\phi},$$

  

where the wavelet matrix, $W_{\phi}$ consists of orthogonal columns of wavelet basis functions, generated by a predefined wavelet basic function, $\phi$, by scaling and translation.

## 2. Image Pyramids and Trees

  

In the previous section, we studied matrix similarity transforms to represent an image in multiple resolutions, using Wavelet transforms. In particular, we have seen that one of the compact dyadic wavelets, called Haar functions, transforms,  an image matrix into a space, spanned by multiresolution basis functions.

  

There is a simple way of representing an image in multiple resolutions, in the space domain, where we define a stack of images, each of which has different resolutions. In the following, we introduce image pyramids and trees for multiresolution image representation.

  

### Image Pyramids

  

Suppose that we have an $N\times N$ image matrix, $f$, where $N=2^J$. We can generate a lower resolution version of this image by down sampling it.

  

Let us form a multi-level pyramid, where at level, $j$, where $j= 0, ….,J$, we have an image, $f_j$ of size , $2^j \times 2^j$. The bottom level of this pyramid consists of the original image, $f=f_{J}$ of size $2^J\times 2^J$ . At level $j$ of the pyramid, we down sample the image to generate a lower resolution image for level, $j-1$. As we keep down sampling, we move towards the top level of the pyramid, until we get a one pixel image, $f_0$.

  

Formally speaking, at level $j$, of the pyramid, we generate an image with resolution $j$ according to,

  

$$\text Resolution (f_{j})= \frac{1}{4}\text Resolution(f_{j+1}),$$

where $j=0,...(J-1).$

  

Reducing the resolution by a factor of $\frac{1}{2}$ means that four pixels of the image at level $j+1$ is replaced by one pixel at level $j$ of the image pyramid. 

**Note:** Storing a pyramid, consisting of images of multiple resolutions is expensive compared to that of Wavelet transform. The storage space required to store the image pyramid is $$O(N^2(1+\frac{1}{4} + \frac{1}{16} + \ldots)) = O(1.33N^2).$$

In general, it is possible to generate image pyramids by a factor of $\frac{1}{K}$, for $K=3, 4, 5, ... $

  

  
  

<figure><p align="center"><img src="https://raw.githubusercontent.com/hazalmogultay/466Images/master/W8/5.png" height=300px><figcaption>Figure 11: Image pyramids generate a stack of images at multiple resolutions with a decreasing order from bottom to the top of the image pyramid. At the bottom of the pyramid we have a full resolution image of $N\times N$, where we assume that $N=2^J$. At the top level, we have a one pixel image. </figcaption></p></figure>

  

**Motivating Question:** When we replace $4$ pixels of level $j$ by a single pixel, at level $j-1$, how do we assign the brightness value of the pixel using the brightness values of the four pixels of the previous level?

  

In order to answer this question, we need an **approximation filter**, which smooths the image prior to down sampling and maps the colors of four pixels at level $j$ to a single color at level $j-1$. There are a variety of approaches for designing an approximation filter. One very simple way is to use an averaging filter. We can, also, use Gaussian or median filters to assign the new color from the old colors.

  

Figure 12, shows a the block diagram representation of  implementation of an image pyramid from an $N\times N$ image by using an approximation filter and down sampling. At each level of the image pyramid, the down sampled image loses some information about the original image. In order to keep the missing information, we can up sample the image and interpolate it to measure the missing information by subtracting the interpolated image from the original image. The difference is called **prediction residual image**. It is possible to reconstruct the original image from its approximated and down sampled version, by adding the prediction residual image to the down sampled image.

  

<figure><p align="center"><img src="https://raw.githubusercontent.com/hazalmogultay/466Images/master/W8/6.png" height=300px><figcaption>

Figure 12: Generation of image pyramids. The upper channel smooths the image by an approximating filter, such as averaging, Gaussian or a median filter. Then, the image is down sampled by a factor of 2 at $x$ and $y$ directions. In order to recover the original image, it is possible to up sample the down sampled image by the same ratio and interpolate the up sampled image to assign a color value to the inserted pixels. However, the recovered image will not be the same as the original image. If we take the difference between the down sampled image and interpolated image, we obtain a difference image, called,  **prediction residual**.  Thus, we need to keep both the down sampled image and the prediction residual to reconstruct the original image, if needed. </figcaption></p></figure>

  

Multiresolution images generated at different levels of the pyramids are independent of each other. Once we down sample an image at level, $j$, we discard the relationship among the pixels of this image with that of the image of level $j-1$. Therefore, we forget about the correspondence of the pixels across the levels, as we move towards the top of the pyramid.



## Image Trees

  

One simple way of relating the images at each level of the pyramid is to use tree data structure. For this purpose, we define a tree, where the nodes at each level correspond to a pixel of images generated at each level of the image pyramid.

  

Mathematically speaking, define a tree,
 $$T= \{P_k, R\},$$ 
 where $P_k$, is the set of nodes at level $k$ and $R$ represents the relation between the parent and children nodes. The set of nodes, $P_k= \{p=(k,i,j)\},$ at level $k$ represents the pixels with coordinate, $(i,j)$.Formally speaking, the image at level $k$ is represented by a set of nodes at that level, as follows;

  

$$P_k =\{p=(k,i,j), k\in [0,(J-1] , \quad (i,j)\in [0,2^k-1]\}.$$

  

Then, we define the relation, $R_k$, between images, $f_k$ and $f_{k-1}$ at level, $k$ and level, $(k-1)$ by relating their pixels corresponding to the parent and its children nodes, as a set of mappings, $R_k= \{R(k,i,j)\}$, where each mapping is defined as;

  

$$R(k,i,j): P_{k-1} \rightarrow P_k.$$

  

In the above formulation, each mapping, $R(k,i,j)$, maps the four pixels of $P_k$, at level $k$ to one pixel at level $k-1$. Therefore, each parent of the tree has four children.

  

The brightness value of each node of the tree, $T$, is obtained by the approximation filter method suggested in the previous section, when we generate an image pyramid.

  

<figure><p align="center"><img src="https://raw.githubusercontent.com/hazalmogultay/466Images/master/W9/pyramide2.png" height=500px><figcaption>Figure 13: An image tree, where each node has four children. The leaves of the tree correspond to the pixels of the image. As we move towards the root, the nodes at each level correspond to the down sampled image.</figcaption></p></figure>

  

It is possible to design image trees, changing the downsampling ratio other than $½$. For example, at each level of the tree we may have more or less than 4 children.

  

There is a wide range of designing schemas for image trees. In the following, we shall explain two of the tree generation methods, namely, **quad trees** and **uniformity trees**.

  


### Quad Trees

  

In many practical problems, we may desire to generate trees, where some of the nodes correspond to regions which may represent objects or part of the objects. These trees enable us to search and browse the content in an image database.

  

From similarity rule of  Gestalt theory, we know that regions in an image correspond to objects if the pixels within a region are similar to each other. In other words, regions of objects are homogenous with respect to a homogeneity predicate, $H$. For example, if a homogeneity predicate is defined as the color similarity, we can merge all the adjacent pixels under the region, $R$, if they have the same color. Then, the homogeneity predicate, $H(R)$ is TRUE. Similarly, we can split a region $R$, if $H(R)$ is FALSE.

  

**Motivating Question:** How can we generate trees, where the nodes at some level consist of homogenous regions, with respect to a homogeneity predicate?

  

Quad tree method generated a tree by **splitting** the image into “homogeneous regions”, in a top down approach. The root of the tree is the image itself. We start from the whole image at the root of the tree and generate **four** children by partitioning it into four disjoint regions. In each region, $R$, the homogeneity predicate is checked. If $H(R)$= TRUE, we keep that region as an end node. Otherwise, we partition the region into four regions to generate four children of a parent node. The procedure continues until all the regions are homogenous.

  

>**Algorithm: Quad tree generation**
>>
>>Input: Image represented by a region, $R$, as the root, homogeneity predicate, $H.$
>>
>>Step 1: Partition a region, $R$ into four disjoint quadrants.
>>
>>Step 2: Check the pixels in each quadrant. If the pixels in a quadrant are homogenous, $H$= TRUE, stop. Otherwise set this quadrant as a region, $R$ and go to step 1.
>>
>>Step 3: Repeat step 1 and 2 until no further regions/ are left to partition.

Region 1 is not homogenous, so we split it into 4 disjoint quadrants and so on… The corresponding quad tree is shown in the right.</figcaption></p></figure>

**Note:** Quad tree algorithm generates large patches for smooth regions of the image and it generates small patches near the object boundaries.

  


### Uniformity Tree (Pinar’s Tree)

  

Consider the satellite image of Figure: 1. Suppose that we can generate an image tree, where the nodes at different levels correspond to buildings, cars, bridges, forests, sea etc. This tree will be very useful for retrieving the objects at multiple resolutions using a tree search algorithm.

  

Pinar Duygulu Sahin suggested a tree generation method, called **[uniformity tree](https://www.sciencedirect.com/science/article/abs/pii/S0031320302002509?casa_token=xLjakJtmQRIAAAAA:k_UFQkdPVUTbosl40j22tJpgFGn5SMpRicPh2pByUEg0QXlWZh8n4vKzgdDmfjrnKHiZ1w0YSQ)**, where the “similar” regions of an image are merged in a bottom up approach.

  

Uniformity tree generates an image at each level by merging the two adjacent regions $R_i$ and $R_j$ if they are “similar” with respect to a **homogeneity predicate**, $H_k(R_i \lor R_j)$ = TRUE. In this approach, at each level, we define a new homogeneity predicate, $H_k$, depending on the **most similar adjacent regions** in that level.

  

At the bottom level of the tree, the collection of **adjacent pixels** with the same color are merged to represent a node. At the next level of the tree, a new homogeneity predicate is defined as,

**“merge two adjacent regions with the closest two colors of that level”** . 


The process continues at each level. The homogeneity predicate is checked to merge two closest colors of adjacent regions. Merging process continues until the root of the tree where there is only one region of a constant color.

  
  

<figure><p align="center"><img src="https://raw.githubusercontent.com/hazalmogultay/466Images/master/W8/8.png" height=500px><figcaption>

Figure 14: Uniformity tree. At the bottom level of the tree the nodes correspond to regions of the same color. The nodes are merged to a parent node, when the corresponding regions are adjacent and the color distance between the two regions is the smallest of all the pairs of the adjacent regions.</figcaption></p></figure>

  
  
  

>**Algorithm: Uniformity Tree Generation**
>>
>>Input: Regions, $\{R_i\}$ with the same color and a color similarity measure.
>>
>>1. For each pair of neighboring regions $R_i$ and $R_j$, compute the color distance $d(R_i,R_j)$ between $R_i$ and $R_j$ according to the selected similarity criteria.
>>
>>2. Insert $d(R_i,R_j)$ into the distance list.
>>
>>3. At each level, identify the minimum of two  colors between the adjacent regions by computing,
>>
>> $$D_j = \min_{\forall ij} D(R_i, R_j) = \min[(r_i - r_j)^2 + (g_i-g_j)^2 + (b_i - b_j)]^{1/2},$$
>>
>> $\quad$ where $r, g, b$ are the brightness values for red green and blue channels.
>>
>>4. Take two neighboring regions $\begin{cases}\text{Merge if }D(R_i,R_j) = D_j \\ \text{Keep otherwise}\end{cases}.$
>>
>>5. Assign the color of the merged pixels as the average color.

  

The above algorithm generates a tree, where at each level the  regions with the "most similar colors" with respect to a similarity measure, are merged. Therefore, at each level the image is represented with a decreasing detail. (See Figure: 15.)

  


  
  
  

<figure><p align="center"><img src="https://raw.githubusercontent.com/hazalmogultay/466Images/master/W8/9.png" height=500px><figcaption>

Figure 15: Outputs of the uniformity tree algorithm at several levels: a) Root (original images), number of regions = 9999; (b) level = 5000; number of regions = 5000; (c) level = 500; number of regions = 500; (d) level = 50; number of regions = 50; (e) level = 10; number of regions = 10; (f) level = 6; number of regions = 6; (g) level = 5; number of regions = 5; (h) level = 4; number of regions = 4; (i) level = 3; number of regions = 3; and (j) level = 2; number of regions = 2.</figcaption></p></figure>

This algorithm generates a large tree with a very high depth. However, the number of tree levels can be reduced  by merging regions, which satisfies 
$$ D(R_i, R_j) = (r_i - r_j)^2 + (g_i-g_j)^2 + (b_i - b_j)^{1/2}\le T,$$ 
at each level, where $T$ is a predefined threshold value.


  


## 3. Sub Band Coding

Recall the filtering techniques in frequency domain, where a filter is designed to pass a predefined frequency band and suppress the rest of the frequencies embedded in an image. For example, a low pass filter outputs the low frequency components of the image, whereas band pass filters pass only the allowed frequency bandwidths and high pass filters pass the high frequency components corresponding to edges and abrupt changes. In short, when we filter an image with a predefined bandwidth, the output image consists only of the frequencies that are passed by the filter. Therefore, one can design a set of filters, called **filter bank**, which decomposes the image into different frequency components.

Sub band coding is the process of decomposing an image into a stack of images, each of which consists of different frequency bandwidths of an image. If the collection of the bandwidth of the filters in the bank covers the entire frequency range, then it is possible to reconstruct the original image from its filtered versions.

**Goal :** Design a set of **collectively exhaustive** filters, called filter bank, such that each filter passes a certain frequency bandwidth and the collection of the filter bandwidths cover the entire spectral band.

The stack of images obtained at the output of a filter bank represents the image content in multiple frequency bandwidths, which enables us to analyze the image content in different frequency ranges, separately.

Sub band coding consists of two types of filter banks:

-  **Analysis Filter Bank** generates a stack of images each of which characterizes the predefined frequency bandwidth of the image.

-  **Synthesis Filter Bank** reconstructs the original image by merging the  stack of images generated at the output of the analysis filter bank.

### Analysis Filter Bank

Analysis filter bank decomposes an image into a set of images each of which is obtained at the output of a predefined filter in the transform domain.

Mathematically speaking, consider the image and its Fourier transform matrix; $f \leftrightarrow \mathcal{F}.$ Let us design a set of filters, called filter bank, $H_i (u,v)$, as follows;

$$H_{n-1}(u,v) = \begin{cases}1 & \text{ for } (u,v)\ge r_{n-1} \\ 0 &\text{ o.w}\end{cases}$$

and for $i= 0,...,n-2$,

$$H_i(u,v) = \begin{cases}1 & \text{ for } r_i\le (u,v)\le r_{i+1} \\ 0 &\text{ o.w}\end{cases},$$

where $r_0 =0$. The number of filters, $n$, depends on the bandwidth, $[r_i, r_{i+1}]$, of each filter, such that the collection of the bandwidths covers the entire spectral band of Fourier transform of the image. In a filter bank, $H_0(u,v)$ is a low pass filter and $H_{n-1}(u,v)$ is a high pass filter. The rest of the filters are band pass filters.

<figure><p align="center"><img src="https://raw.githubusercontent.com/hazalmogultay/466Images/master/W9/subband.png" height=200px><figcaption>

  

Figure 16: The schematic representation of an image, $f$ in space domain, its amplitude $|F(u,v)|$ and phase $\angle F(u,v)$ spectrum. In this particular example, the amplitude spectrum is splitted into four bands using ideal filters. Thus, a filter bank of four filters $H_i, \quad \text for \quad i =0,1, 2,3$ is defined. These filters cover the entire spectral band of the image in frequency domain. </figcaption></p></figure>

Figure 16 shows a sample decomposition of the image into a set of frequency bandwidths using ideal filters. In this decomposition, $H_0(u,v)$ is a low pass filter, which passes the low frequencies which is less then a predefined frequency. $H_1(u,v)$ and $H_2(u,v)$ are the band pass filters. $H_3(u,v)$ is a high pass filter.




  
  

<figure><p align="center"><img src="https://raw.githubusercontent.com/hazalmogultay/466Images/master/W9/Hu.png" height=300px><figcaption>

Figure 17: The profile of an ideal band pass filter, defined in a filter bank. The frequency bandwidth of the filter, $H_i$ is defined in the range of $[r_i, r_{i+1}]$</figcaption></p></figure>

  

Figure 17 shows the profile of one of the an ideal filter in the bank. Depending ofn the application domain there are total of $n$ filters in the bank. When we filter the image in Fourier domain with the filter bank, we generate $n$ images as follows;

  

$$F_i(u,v) = F(u,v)\cdot H_i(u,v), \quad \text for \quad i= 0,...,n-1$$

  

This process decomposes the image into a set of band-limited components, called sub bands. Inverse Fourier transforms generate a stack of images in different frequency content as follows,

  

$$f_i (x,y) \rightarrow F_i(u,v).$$

  

Therefore, the filter bank decomposes an image into a set of images with different frequency content, called the **sub bands of images**.

**Note**: If the filters in the bank are ideal, there is no need to overlap the bandwidths. However, as we have seen in the previous chapters the discontinuity of the ideal filters results in some undesired effects at the output. Thus, it is desirable to use different filters, such as,  Gaussian or Butterworth filters. In this case, we allow overlaps between the bandwidths of the filters to cover the entire spectrum.





### Synthesis Filter Bank

Synthesis filter bank generates the original image from the output of the analysis filter bank.

Suppose that we are given  a set of images obtained at the output of the analysis filter bank, using a set of ideal low pass filters, as follows;

$$f_i (x,y) \rightarrow F_i(u,v), \quad \text{for} \quad i= 0,....(n-1),$$

If we generate the analysis filter bank using a set of mutually exclusive and collectively exhaustive ideal low pass filters, then, it is possible to reconstruct the original image in the frequency domain  by adding all the sub band images as,

$$F (u,v)= \sum_{i =0}^{n-1} F_i(u,v),$$

Unfortunately, due to many practical reasons the filters in the bank consists of overlapping bandwidths. Furthermore, since the images in the analysis filter bank is very sparse, we store them after down sampling them. This may result in a set of compressed image with a tradeoff in information loss. Therefore, the synthesis filter bank should be designed to  compensate for the losses and overlaps.

In the following sections, we design the analysis and synthesis filter banks of   a one and two dimensional images, where we have only a low pass and its complementing high pass filter in the bank.






### Two-Band Coding of One-Dimensional Images

  

Let's  decompose a one dimensional image function, $f(x)$, into two bands with only two  filters, namely a low pass and its complementing high pass filter with a predefined cutoff frequency . The cutoff frequency of the filters depends on the application domain. At the output of the filters, we obtain a low pass filtered image, called, **approximation part** and a high pass filtered image, called **detail part**.

  
  

<figure><p align="center"><img src="https://raw.githubusercontent.com/hazalmogultay/466Images/master/W9/h0h1.png" height=250px><figcaption>

Figure 18: Filter bank for two band coding. $H_0(u)$ is a low pass filter with the cutoff frequency, $r_0$ and $H_1(u)$ is a high pass filter with the cutoff frequency, $r_1.$

</figcaption></p></figure>

  

Let, the Fourier transform pair of a one dimensional image is $f(x) \leftrightarrow F(u)$. Then, two-band coding involves the following analysis and synthesis techniques.

  

**1. Analysis Filter Bank: Two-band Coding of One dimensional Images**

At the output of the analysis filter bank, we decompose the original image into two images, called the **approximation part** and **the detail part**. The approximation part is obtained at the output of the low pass filter. Thus, it is the blurred version of the image. The detail part   is obtained at the output of the complementing high pass filter. Thus, it consists of the high frequency components, corresponding to edges. The following algorithm, explains the basic steps of the Analysis Filter bank.

**Algorithm: Analysis Filter Bank for Two-Band Coding of One Dimensional Images**

  

-  Step 1: Design a low pass, $h_0(x) \leftrightarrow H_0(u)$ and a high pass filter, $h_1(x) \leftrightarrow H_1(u)$, which cover the entire spectral band of the image, 

$$f(x) \leftrightarrow F(u).$$

-  Step 2: Find,

$$f_i(x) = f(x)*h_i(x)\leftrightarrow F_i(u) =F(u).H_i(u),$$ 

for $i= 0$ and $1$, for the approximation part and the detail part of the original image.

-  Step 3: Since the image at each band,
$$f_i(x) \leftrightarrow F_i(u)$$ 

is very sparse, there is no need to keep the dimension of the original  image. Thus, we down-sample the output of the filter, $f_i(x) \rightarrow F_i(v)$ by two to obtain the approximation part, $f_{0 \downarrow}$ and the detail part, $f_{1 \downarrow}$.


<figure><p align="center"><img src="https://raw.githubusercontent.com/hazalmogultay/466Images/master/W9/f1.png" height=250px><figcaption>

Figure 19: Sub band coding of a one dimensional image $f(x)$. At the output, we have two images, the approximation part, $f_{0 \downarrow}$ and the detail part, $f_{1 \downarrow}$. Note: $\boxed{2 \downarrow}$ indicates down sampling the image by a factor of two.</figcaption></p></figure>

The approximation part, $f_{0 }$ is a smoothed version of the image whereas the  detail part, $f_{1}$ consists of the sharp changes. Both approximation and real part have size $N/2\times N/2$, due to down sampling.


**Motivating Question:** Where do we use the approximation and detail parts obtained at the output of sub band coding?

Two-band coding decomposes the image into approximation part, which represents the low frequency components and the detail part, which represents the high frequency components of the image. This representation allows us to enhance the high and low frequency components separately. It is, also, very useful for image compression.



**2. Synthesis Filter Bank: Two-Band Decoding of One Dimensional İmages**

In order to decode the original one dimensional image $f(x)$ from its approximation and detail parts, we reverse the operations defined above. At the first step, we need to up sample the approximation and detail parts. Then, we need to design reconstruction filters, $g_0(x)$ and $g_1(x)$ to recover the image from the approximation and detail parts. There are many ways of designing the reconstruction filters. It can be shown that for a one dimensional two-band coding, the analysis and synthesis filters are related to each other by the following equations;

$$g_0(x) = (-1)^xh_1(x)$$ 
and

$$g_1(x) = (-1)^{x+1}h_0(x)$$

Designing the reconstruction filter is beyond the scope of this course. The interested reader can refer to [https://www.ece.uvic.ca/~imanmoaz/homepage/files/report_filter_bank.pdf](https://www.ece.uvic.ca/~imanmoaz/homepage/files/report_filter_bank.pdf) .

The brief algorithm of the synthesis filter bank for two band decoding of the one dimensional images is given below. 

**Algorithm: Synthesis Filter Bank for Two-Band Decoding of One Dimensional Images**
  

-  Step 1: Up sample the the images, $f_0(x)$ and $f_1(x)$, respectively, to put the image into its original sizes.

-  Step 2: Design the synthesis filters using the the low pass and high pass filters of the analysis bank,

$$g_0(x) = (-1)^xh_1(x)$$ 

and

$$g_1(x) = (-1)^{x+1}h_0(x)$$ 

- Step 3:Filter the up sampled images,

$$f_0(x)*g_0(x)$$ 
and

$$f_1(x)*g_1(x)$$

-  Step 4: Add the outputs of the synthesis filters,

$$f(x) = f_0(x)*g_0(x) + f_1(x)*g_1(x).$$

The above algorithm for two band decoding does not reconstruct the original image from its analysis filter bank if the overlap between the high pass and low pass filters are significantly large. Also, during the up sampling process we may loose some information about the original image.

### Two-Band Coding of Two-Dimensional Images

The above approach to decompose a one dimensional image into two bands, can be extended to a two-dimensional image function, $f(x,y)$.

At this time, we consider each column and each row of an image as a one dimensional discrete function and adopt the analysis and synthesis filter bank algorithms, suggested in the previous section.  Thus, we apply one dimensional low pass and high pass filters to the  rows and columns of the two dimensional image, separately. In order to decompose the rows of the image, we define, a one-dimensional low pass filter, 
$$h_0(x)\leftrightarrow H_0(u)$$ 
and a one dimensional high pass filter,
$$h_1 (x)\leftrightarrow H_1(u).$$

Similarly, in order to decompose the columns of the image, we define a one-dimensional low pass filter, 
$$h_0(y)\leftrightarrow H_0(v)$$ 

and a one dimensional high pass filter,

$$h_1 (y)\leftrightarrow H_1(v).$$

Then, we filter each row and column of an image by the above one dimensional filters in transform domain and down sample the filter outputs. Finally, we merge the one dimensional outputs in $x$ and $y$ directions to obtain two dimensional outputs. 

The steps for two band coding, shown in Figure 6, are explained in the following algorithm.

**Algorithm: Analysis Filter Bank for Two-band Coding of Two Dimensional Images**

-  Step 1: Low pass filter the rows of  f(x,y) and down sample the filtered rows to generate  $f_0(x)$.
- Step 2: High  pass filter the rows of  f(x,y) and down sample the filtered rows to generate  $f_1(x)$.
- Step 3: Low pass filter the columns of  f(x,y) and down sample the filtered columns to generate $f_0(y)$.  

- Step 4: High pass filter the columns of  f(x,y) and down sample the filtered columns to generate $f_1(y)$. 


- Step 3: Merge the filtered and down sampled rows and columns,

    - $f_0(x)$ and $f_0(y)$ to generate two dimensional approximation part, $a_0(x,y)$

    - $f_0(x)$ and  $f_1(y)$ to generate two dimensional horizontal details, $d_v(x,y)$

    - $f_1(x)$ and $f_0(y)$ to generate two dimensional vertical details, $d_h(xy)$
    
    - $f_1(x)$ and columns, $f_1(y)$ to generate two dimensional diagonal details, $d_D(x,y).$

    The above analysis filter bank algorithm decomposes a two dimensional image into four down sampled images, representing different frequency components. Design of the filters depends on the application domain.

# Chapter Conclusion

Can we represent  various levels of details of an image at the same time? Can we design a matrix similarity transform, which maps an image into a space, spanned by functions with multiple scales? Can we combine images at different resolutions under the same tree data structure? Can we use Fourier transform to decompose an image into several frequency components?

The answers behind the above questions lie in multiresolution image analysis. In this chapter, we study three major techniques for decomposing an image into its various components of scale, resolution and frequency bandwidths. A very important technique is wavelet transforms, where we define a real, finite duration  and admissible  basic wavelet function. Then, we generate a set of orthogonal basis functions by translating and scaling the basic wavelet function. When we represent image as a matrix, it is possible to find a matrix similarity transform, which maps an image from the space domain to a wavelet domain, as in the Fourier, Cosine and Hadamard transforms.

The second group of techniques involve representing an image in a tree data structure, where each node correspond to a pixel or collection of pixels, which represents a homogenous region with respect to a homogeneity predicate. In this representation, each level of the tree corresponds to an image with different coarseness of the same image.

Finally, we study a technique, called sub band coding. In this technique, we use the power of Fourier transform, to represent various frequency components of an  image, as a stack of images. Each image in the stack is generated  by filtering the original image in a different frequency bandwidth.  The filters, called filter bank, covers the entire frequency spectrum of the image from low pass, band pass and high pass ranges.
