## 1. Image Representation by Functions

Let us start by defining a function.

Loosely speaking, a function  can be defined as a   relation between two sets, $D$ and $R$.  It relates each element in $D$ to an element in $R.$ The set, $D$, is called the **domain** and the set, $R$, is called the **range** of the **function** $f$, respectively.

A **continuous image**, sometimes called **analog image**, can be represented by a continuous function, where both domain and range of the function is continuous. It is indicated by,

$$f : D\rightarrow R.$$  

The elements in the domain of this function, $d\in R^2$ , consist of 2-dimensional coordinates, $d=(x,y)$ and the elements of the range, $r$, consist of the brightness values, $l \in R$.

A **motion picture** is represented by a function $f(x,y,t)$ , where the domain includes the time in addition to the coordinates of brightness values, yielding three dimensional elements, $d=(x,y,t)$, in domain $D$. The range of the function, $l \in R$, remains the same.

A **discrete image** can be represented by a discrete function. In discrete images, the domain, $D$, of the function

$$f: D\rightarrow R$$

becomes a discrete grid of pixel coordinates and the range , $R$, where $l \in R$, is the same as an analog image, which consists of continuous brightness values.

A **digital image** can be represented by a function, where the domain, $D$ and range $R$ of the function,

$$f: D\rightarrow R,$$

consists of only integer values. Digital image representation enables us to represent the pixel coordinates in a two dimensional finite grid of integers. The brightness value of each pixel, represented by $k$ bits, is represented by an integer value, selected  in an interval $[0, (2^k-1)]$.
 

In the context of this course, we shall only work with digital images.

Depending on the type of the image the dynamic range and dimension of the domain and range varies, as described below:


### Gray Scale Image

A digital gray scale image is represented by a function

$$f: D\rightarrow R,$$

where the domain, $D$, is a two dimensional grid of integers and the range, $R$, has elements, $l \in [0,L-1]$, having $L = 2^k$ distinct brightness values. The value $0$ represents the black color, while the value $(L-1)$ represents the white color. For an image, represented by $k=8$ bits/ pixel, there are a total of 256 gray tones between black and white.


### Color Image

A digital color image is represented by a set of functions,

$$f_i: D\rightarrow R,$$

where the index, $\quad i=R,G,B$, generates 3 different discrete functions, $f_i$. The indexes, $R$, $G$ and $B$ are three [primary colors](https://en.wikipedia.org/wiki/Primary_color) for red, green and blue, respectively. Thus, each pixel of a color image is represented by a three dimensional vector, which shows the intensity values of the primary colors.

### Multispectral and Hyperspectral Image

Multispectral images are the extension of digital color images, where the number of representation functions,

$$f_i : D\rightarrow R$, = $f:(x,y) ,\quad i=1,...,I ,$$

depends on the number of spectral bands, $I$. There are 4-10 spectral bands in **multispectral** images. When the number of bands exceeds 10, we call it **hyperspectral image**. A typical hyperspectral image may consist of around $I=100$ or more narrow spectral channels. In this case, the image of the same scene is represented by a set of $I$ functions. In this representation, an image becomes a discrete volume of size $N\times M\times I$, where $N\times M$ is the image size and $I$ is the number of spectral bands.

### Binary image

  

  

In a binary image, the domain is again a two dimensional grid of integers. However, the range , $l \in [0,1]$, consists of 0 corresponding to black and 1, corresponding to white.

  

  




## 2. Image Representation by Arrays and Matrices

Digital images can be represented by arrays, matrices or vectors. Let's refresh our memories about the basic definitions of these simple mathematical structures:

An **array** is a [data structure](https://en.wikipedia.org/wiki/Data_structure) consisting of a collection of elements each identified by at least one array index or key.

A **matrix** is a two dimensional array that has a size $N\times M$, where N and M are nonnegative integers, showing the number of rows and columns, respectively.

A **vector** is a one dimensional array.

An array may have any number of dimensions. It can, also, be 2-dimensional, scalar or empty. Matrices and vectors are just convenient names given to a particular subset of arrays.

**Note:** Array operations and matrix operations are different. Array operations are arithmetic operations where element wise addition, multiplication and division is applied. Matrix addition is the same as array addition. However, matrix multiplication and inverse is different.

**Q:** Refresh your Linear Algebra for matrix multiplication and inverse.

Consider a 2- dimensional digital image with multiple bands. This image can be represented by **matrix, $f_i$** of size an $N\times M$ for each band $i$, as follows;

$$f_{i} = \begin{bmatrix}f_i(0,0) \qquad \quad& ... & f_i(0,N-1) \\ && \\f_i(M-1,0) & ... & f_i(M-1,N-1)\end{bmatrix}.$$

In the above matrix representation of a digital image for channel $i$, each entry


$$f_{i}(x,y) \in \ [0,L-1],$$

represents a brightness value at coordinate $(x,y)$ in the interval of $[0, L-1]$ . If we have total of $L$ brightness values, we need

$$\ L=2^k,$$


which means that we need $k$ bits/pixel. The interval, $[0,L-1]$ is called the **range of the color band.**

The space required to store $i$th band of this image is

$$S_{i} = N\times M\times k .$$

If we have $i=1,......,I$, total of $I$ channels, then the requires storage space is

$$S = N \times M \times k \times I.$$

If we have motion picture image

$$S = N \times M \times k \times T ,$$

where $T$ represents the number of frames.

**Fact:** The above simple arithmetic shows that storing images is expensive 😊

In order to obtain a **vector representation of an image**, all we have to do is to concatenate the rows of a matrix one after the other.

$$ f = [f(0,0),....f(0, N-1),..., f(M-1,N-1)].$$

This way, we obtain an $N \times M$-dimensional vector for each band. In fact, we may concatenate all the bands to obtain a vector of size $N \times M \times I$.

However, we have to keep in mind that processing a vector in such a high dimensional vector space brings many problems.

  



# A Simple Image Formation Model

Our Human Visual System forms an image of an object by capturing the brightness values reflected from the objects by light sensitive sensors, namely rods and cones. Similarly, a digital image is formed by capturing the brightness values reflected from an object and converting it into a set of integers.

There are two external components, which affect the formation of an image, in an image acquisition device:

1. [Illumination](https://en.wikipedia.org/wiki/Illumination_%28image%29) , measured by the unit, [$i(x,y)$  $lambert/m^2$](https://en.wikipedia.org/wiki/Lambert_%28unit%29)

2. [Reflectance](https://en.wikipedia.org/wiki/Reflectance) or transmission (or transmission for x-ray) parameter for x-ray, 
$$0 \leq r(x,y) \leq 1.$$

  

  

<figure><p align="center"><img src="https://raw.githubusercontent.com/hazalmogultay/466Images/master/W3/L3-f3.jpg" height=300px><figcaption>

Figure 3: A simple image formation model, first suggested by **[Hassan Ibn Haytham.](https://en.wikipedia.org/wiki/Ibn_al-Haytham)**</figcaption></p></figure>

  

  

Figure 3 shows simple image formation models. Note that there are two ways of forming an image: The first one is by **reflection** , as in optical camera images. The second one is by **transmission**, as in X-ray images.


An image is formed as a function by multiplying the illumination and reflectance or transmission parameters, as follows,

$$f(x,y) = i(x,y) r(x,y) = l.$$

The range of this function is $L_{min}\leq f(x,y) \leq L_{max},$

where $L_{min}$ and $L_{max}$ depends on the properties of the image acquisition device.

In order to generate a digital image, where both the domain and range consists of integer values, we normalize the physical domain and range. The physical domain is normalized by generating a two-dimensional discrete grid of size $N \times M$, with coordinates, $x\in[0,N-1]$ and $y\in[0,M-1]$ of the digital image function $f(x,y)$.

The range of the image is normalized by shifting the physical brightness values, as follows,

$$\left.\begin{array}{cc}L_{min} \rightarrow 0 \\L_{max} \rightarrow L-1\end{array}\right\}L \ \epsilon \ [0,L-1] \\. $$

Thus, the output of a digital image acquisition device is an $N \times M$ image, where each integer valued coordinate, $(x,y)$, correspond to a picture element, called **pixel, p** with integer **brightness value $l$**, at each spectral [bandwidth](https://en.wikipedia.org/wiki/Bandwidth_%28signal_processing%29).

## Sampling and Down-Sampling

Sampling a continuous image digitizes the domain of the image function, $f$. Recall that this function can also be represented by an $N \times M$ matrix, for each color band. Sampling methods are based on the famous **Sampling Theorem** of [**Claude Shannon**](https://en.wikipedia.org/wiki/Claude_Shannon). We shall explore this theorem and sampling methods later.

# Image Interpolation

In some image processing problems, we may need to reconstruct the continuous image from its sampled and quantized version or we may need to smooth the digital image as if it looks continuous. For example: When we  **zoom in** a certain location of a digital image or we  **rotate** and **stretch** an object, the appearance of the image can get distorted. Zooming process enlarges the physical size of pixels, resulting in fat bits.   In order to obtain a smooth image, we insert extra pixels in between the existing pixels. At this point, we need to fill in the blank pixels after a zooming operation. Similar problems occur when we rotate or stretch the objects in an image.


The techniques, which are used to obtain a continuous image from its digital version is called **interpolation** or **resampling**.

  
The techniques to improve the resolution of a down-sampled image is called **up-sampling.**



Interpolation and up-sampling are classical problems, in numerical analysis, where we estimate the values at each unknown point, using the values of available data.

There are many techniques to interpolate a two dimensional digital function, which represent an image. Among them, we suffice to introduce 3 simple techniques as summarized below:

  

  



### 1. Nearest Neighbor Interpolation

  

Nearest neighbor interpolation fills in the brightness values of blank pixels according to the nearest neighbor rule.

  

**Nearest Neighbor rule:** Assign the intensity value of the nearest pixel as the intensity value of a blank pixel.

  

**Note:** Depending on the neighborhood system “nearest neighbors” may change and you may have multiple nearest neighbors. In this case, you need to revise the above nearest neighbor rule for a particular image dataset.

### 2. Bilinear Interpolation

Bilinear interpolation employs 4-nearest neighbors to estimate the intensity of a blank pixel. Suppose that the unknown brightness value of the center pixel is represented in terms of the following equation,

$$v(x,y) = ax+by+cxy+d,$$

where $(x,y)$ is the spatial coordinates of pixels and $a$, $b$, $c$, $d$ are the unknown parameters to be estimated by using the known brightness values of the neighboring pixels. The 4 unknown parameters are solved by using four equations obtained by using the brightness values of 4 neighboring pixels.

### 3. Bi-cubic Interpolation

Bi-cubic interpolation is just a simple extension of bilinear approximation.This time, we employ 16 nearest neighbors to estimate the intensity of an unknown value of a pixel. The interpolation equation is formed as follows;

$$v(x,y) = \sum_{i=0}^{3} \sum_{j=0}^{3} a_{ij} x^i y^j. $$

Notice that, in the above equation there are a total of 16 unknown parameters, $\{a_{ij}\}, i,j=0, 1, 2, 3$. These parameters are calculated from brightness values of 16 known values of neighbors of an unknown pixel. Obviously, this approach yields a smoother fit to the unknown pixel with a substantial computational cost.

# Neighborhood and Distance
Recall that a digital image is represented by a digital function lying on a discrete grid. Since both the domain and range of this function is integer valued, some of the concepts of classical calculus, such as distance between two points or derivative are not directly applicable. Recall that a digital image is represented by a digital function lying on a discrete grid. Since both the domain and range of this function is integer valued, some of the concepts of classical calculus, such as distance between two points or derivatives are not directly applicable. Furthermore, for processing an image in a discrete grid, we need to define additional concepts, such as, **neighborhood, adjacency or connectivity.** 

In the following, we make some useful definitions, which make manipulation of images possible.

## Neighborhood

Neighborhood is an important concept in discrete geometry. The neighbors of a pixel are defined according to the neighborhood system.

**4-Neighborhood System:** Consider a pixel, $p$, with brightness value $l=f(x,y)$ at location $(x,y)$. 4-neighbors of pixel, $p$, indicated by $N_4(p)$, is defined as the set of pixels at locations,

$$N_4(p) = \{(x-1,y), (x, y-1), (x+1,y), (x, y+1)\}.$$

A pixel $q\in N_4(p)$ is called the neighbor of $p$.

<figure><p align="center"><img src="https://raw.githubusercontent.com/hazalmogultay/466Images/master/W3/L3-f12.jpg" height=150px><figcaption> 

Figure 12.a: In the 4-Neighborhood system, a pixel, $p$, has **up, down, left** and **right** pixels as neighbors. The distance between pixel, $p$, and its neighboring pixel, $q$ is assumed to be 1 pixel. </figcaption></p></figure>

**Diagonal Neighborhood System:** Diagonal neighbors of the  pixel, $p$, is defined as the set,
 

$$N_D(p)= \{(x-1,y-1), (x+1, y-1), (x-1,y+1), (x+1, y+1)\}.$$


**8-Neighborhood System:** A pixel, $q$, is said to be in the 8-neighborhood of the pixel, $p$, if

$$q\in N_4(p)+N_D(p).$$

Note that the 8-neighborhood system covers all of the surrounding pixels of $p$, in a discrete grid.

It is possible to define higher order neighborhood systems. For example, a 24-neighborhood system includes additional pixels to $N_8(p)$ as,

## Distance and Similarity

>**Q1:** Considering the fact that digital images are arrays of integer values, lying on an integer valued grid, how do we measure distance between two pixels?  

>**Q2**: How do we measure similarities between two images? For example, suppose that we are given images of bananas, pears and apples. Which ones are more similar? Bananas and pears or bananas and apples?

These are tough questions and there are no unique answers for them.

**Distance**, sometimes called **similarity**, is a very philosophical and subjective concept, which depends heavily on the application domains and goals. An interesting assay can be found in ([https://medium.com/x8-the-ai-community/understanding-similarity-measures-in-ml-33deb0bf094](https://medium.com/x8-the-ai-community/understanding-similarity-measures-in-ml-33deb0bf094) )

Mathematically speaking, distance, $D(p,q)$, between two elements, $p$ and $q$ or $z$, is a **metric** if and only if it satisfies the following conditions:

1. $D(p,q) \ge 0$ and $D(p,q) = 0 \Leftrightarrow p=q,$
2. $D(p,q) =D(q,p),$
3. $D(p,z) \le D(p,q) + D(q,z).$

We can define two types of distances in Image processing:
- Spatial distance, between two pixels,
- Color distance, between two colors.

  


### 1. Spatial Distance Between  Pixels

Spatial distance is defined as the distance between two pixels.

If we had a **continuous image**, we can use the classical distance metrics to measure the distance between two points. For example, we could measure the **Euclidean distance** between two pixels $p_1$ and $p_2$, as follows,

$$D_{e}(p_1,p_2) = [(x_1-x_2)^{2}+(y_1-y_2)^{2}]^{1/2},$$

where the coordinates of the pixels are defined as,
$$p_1=(x_1,y_1), \ p_2=(x_2,y_2).$$ 

Recall that for continuous images,
$$(x,y) \ \epsilon \ R^{2}.$$

The definition of distance depends on the predefined neighborhood system for an image dataset. Among them the most widely used ones are **city block** and **chessboard** distances:

$$D_{4}(p_1,p_2)= |x_1-x_2| + |y_1-y_2|,$$

where pixel coordinates for $p_1$ and $p_2$ are $(x_1,y_1)$ and $(x_2,y_2)$, respectively.

# Discrete Geometry

In this section, we define  the concepts of **adjacency, path, connectivity, region, boundary and edge** on a discrete grid.

## 1-Adjacency

Let, $V$ be a set of pixels defined over the intensity values used to define adjacency.

For example; in a 2 bits/pixel image, we are allowed to have four colors with the color codes 0, 1, 2, 3. When we define $V= \{0,1\}$, this set consists of pixels with gray values 0 and 1 only.

For example, given a binary images, the set $V = \{1\}$ set of white pixels may correspond to abject and $V = \{0\}$ is the set of black pixels may correspond to background.

The concept of adjacency depends on the definition of the neighborhood system:

If we are working with a 4-neighborhood system, we define 4-adjacency as follows:

**4-Adjacency:** Pixels, $p(V)$ and $q(V)$, defined over the intensity values set $V$, are 4-adjacent, if $$p(V)\in N_{4}(q(V)).$$

If we are working with an 8-neighborhood system, we define 8-adjacency as follows:

**8-Adjacency:** Pixels, $p(V)$ and $q(V)$, defined over the intensity values set $V$, are 8-adjacent, if $$p(V) \in N_{8}(q(V)).$$

## 2. Path



Path between two pixels, $p(V)$ and $q(V)$ is a sequence of pixels defined as follows,

$$\underbrace{(x_{0},y_{0})}_{\text{p}} (x_{1},y_{1}) ... \underbrace{(x_{n},y_{n})}_{\text{q}},$$

where $(x_{i},y_{i}) \in N_k(x_{i+1},y_{i+1})$ and $k$ indicates the neighborhood system.

  

  



## 3. Closed path

If the pixels at locations $(x_{0},y_{0})=(x_{n},y_{n})$ in a path, then the path is closed.

**Note:** Both the definition of path and closed path depends on the neighborhood system, defined for an image dataset.

  

## 4. Connectivity

Let, $S(V)$ be a set of pixels, with the brightness values defined over the set $V$, in an image, the pixels $p(V)$ and $q(V)$ are **connected** in $S(V)$ if there exists ($\exists$) a path between them consisting entirely of pixels in $S(V)$.

## 5. Connected Component

Let $p\in \ S(V)$ be a pixel, the set of all pixels, $q(V)$, which are connected to $p$ , is called the connected component.

## 6. Connected Set

If $S(V)$ has only one connected component, with respect to a set $V$, then it is a **connected set** with respect to $V$.

## 7. Region

Let $R$ be a subset of pixels. $R$ is called a **region**, if it is a connected set with respect to a predefined set $V$.

## 8. Region Adjacency and Objects

Two regions, $R_{i}$ and $R_{j}$, are adjacent if $R_{i} \cup R_{j}$ is a connected set, with respect to a predefined set $V.$

**Fact: Objects** can be described by **regions**.

## 9. Region Complement and Background

Given an image with $\{R_{i}\}_{i=1}^{n}$ with $N$ object , the complement of the regions $R^{c}$ is called **background.**

## 10. Holes


The region in the complement of $R^{c}$, which is surrounded by the pixels $p \in R_{i}$ is called a hole.

  

  



## 11. Boundary

Intuitively speaking, **boundary**, sometimes called **border** or **contour**, is defined by a set of pixels surrounding a region, which may represent an object. However, since we define boundary in a discrete grid, we need to be more specific about the definition of boundary.

 

Mathematically speaking, there are two types of boundaries:

 
**1. Inner boundary** of a region $R$ is a set of connected pixel $p \in R$, where $p \in N_{k}(R^{c}).$

**Note:** Inner boundary pixels lie in the object.

**2. Outer boundary** of a region $R$ is a set of connected pixels, $p \in R^{c}$, where $p \in N_k(R)$.

**Note:** Outer boundary pixels lie outside the object.

## 1. Linear Operations

Given an input image $f(x,y)$, an operator, $H$ generates an output image $g(x,y) = H[f(x,y)]$.

$H$ is a linear operator if [superposition](https://en.wikipedia.org/wiki/Superposition_principle) holds:

$$H[a_{1} f_{1}(x,y) + a_{2} f_{2}(x,y) ] = a_{1} H[f_{1}(x,y) ] + a_{2} H[f_{2}(x,y) ].$$


## 2. Non-linear Operations

If the superposition does not hold, then the operator $H$ is called [nonlinear](https://en.wikipedia.org/w/index.php?title=Nonlinear_operator&redirect=no).

**Exercise:** Show that **max, min, absolute value**  are nonlinear operators. In other words, given two images, $f_1(x,y)$ and $f_2(x,y)$, the superposition property does not hold, when these operations are applied on the image functions.

**Answer:** Let us evaluate the superposition property of **max, min** and **abs** operators:

\begin{align*}
max(a_{1} f_{1}(x,y)+ a_{2} f_{2}(x,y)) &\neq a_{1} maxf_{1}(x,y) + a_{2}max f_{2}(x,y).\\
min(a_{1} f_{1}(x,y)+ a_{2} f_{2}(x,y)) &\neq a_{1} minf_{1} (x,y) + a_{2}min f_{2}(x,y).\\
abs(a_{1} f_{1}(x,y)+ a_{2} f_{2}(x,y)) &\neq a_{1} abs (f_{1}(x,y)) + a_{2}abs( f_{2}(x,y)).\\
\end{align*}


Since the superposition property does not hold, non of these operators are linear.

    f_{1}(x,y) &\div f_{2}(x,y)\\
\end{align*}

## 4. Matrix and Vector Operations

Although vectors and matrices are considered as one and two-dimensional arrays, respectively, vector and matrix operations are different from the array (arithmetic) operations. Given two images $f(x,y)$ and $g(x,y)$, as,

$$F =f(x,y) = \begin{bmatrix} f(0,0) & f(0,1) \\ f(1,0) & f(1,1) \end{bmatrix} \ and \ G =g(x,y) = \begin{bmatrix} g(0,0) & g(0,1) \\ g(1,0) & g(1,1)\end{bmatrix}$$



Matrix multiplication of F and G gives;



\begin{align*}
    F.G &= \begin{bmatrix} f(0,0) & f(0,1) \\ f(1,0) & f(1,1) \end{bmatrix}\begin{bmatrix} g(0,0) & g(0,1) \\ g(1,0) & g(1,1) \end{bmatrix}\\
    &= \begin{bmatrix} f(0,0)g(0,0)+f(0,1)g(1,0) & f(0,0)g(0,1)+f(0,1)g(1,1) \\ f(1,0)g(0,0)+f(1,1)g(1,0) & f(1,0)g(0,1)+f(1,1)g(1,1) \end{bmatrix}.
\end{align*}

## 5. Probabilistic Operations
When there is uncertainty about outcomes of an event, we represent this event by a [random variable ]([https://en.wikipedia.org/wiki/Random_variable](https://en.wikipedia.org/wiki/Random_variable)), $z_k$. The model for this event cannot be described by a deterministic function, because there are uncertainties. Instead, it is described by a [probability density function](https://en.wikipedia.org/wiki/Probability_density_function), $p(z_k)$.

A probabilistic event is, then, represented by two tuples $[z_k, p(z_k)]$. Once we define the random variable and its associated probability density function, we can use probabilistic operations to process images.

**Note:** Since we are dealing with digital images, in the above definitions, we assumed that the random variable is discrete. When the random variable is discrete, instead of using the term “probability density function”, we may use the term **probability mass function**.

Probabilistic operations include the **mean value**,

$$\mu = E(z_i)=\sum z_i p(z_i) ,$$

**variance**,

$$\sigma^2 = E(z_i-\mu)^2= \sum (z_i - \mu ) p(z_i) ,$$

or in general, **$k^{th}$ central moment**

$$\mu^k = E((z_i-\mu)^k) = \sum (z_i - \mu )^p p(z_i) ,$$

where $E$ indicates the expected value operator. The random variable $z_i$ shows all possible outcomes related to a probabilistic event, represented by the probability density function $p(z_i).$ The summations, in the above equations, are taken over all possible values of the variable $z_i$.

**Q:** How do we define some random events concerning image processing?

There are many ways of defining random events in image processing. Two popular ways are the followings:

- We can assume that the images are generated by a random source,

- We can assume that the pixel brightness values are generated by a random source.

Let's give some examples for the above cases.

**Ex:** [Hubble telescope](https://en.wikipedia.org/wiki/Hubble_Space_Telescope) generates images of the *same part of the space* many times. Assume that each image $f_i(x,y)$, as an outcome of a random source. We may define the brightness value of each pixel at location (x,y) as a random variable, for all the images, of the same scene, $f_i(x,y)$, generated by Hubble. The probability density function for each pixel can be estimated by using the pixel brightness values $z_i(x,y)$, by roughly

$$ p(z_i(x,y)) \approx { {n_i(x,y)}}/{ n},$$

where $n_i(x,y)$ is the number of brightness values with $z_i(x,y)$, for the pixel at location $(x,y)$ and $n$ is the total number of images captured for the same scene. We can use this discrete function as a **probability mass function**, which is the discrete form of the probability density function.

However, if we need a continuous probability density function, then, we can define a continuous random variable , $z$ and its associated probability density function $p(z)$. Then, we can fit this rough discrete function to one of the well-known analytic forms, such as [**Gaussian**](https://en.wikipedia.org/wiki/Normal_distribution), which has the following analytical form:

$$p(z) = \frac{1}{\sqrt{2\pi r}} exp[ -\frac{1}{2r^2} (z - \mu) ]$$

by some estimation methods, which is beyond the scope of this course (If you are interested, you can visit [https://towardsdatascience.com/maximum-likelihood-estimation-explained-normal-distribution-6207b322e47f](https://towardsdatascience.com/maximum-likelihood-estimation-explained-normal-distribution-6207b322e47f)) .
<figure><p align="center"><img src="https://raw.githubusercontent.com/hazalmogultay/466Images/master/W3/image24.png" height=300px><figcaption>Figure 25: Univariate Gaussian distribution with mean value $\mu$, and standard deviation $\sigma$. </figcaption></p></figure>

  
  
  

Once we estimate the analytical forms of the probability density functions for each pixel brightness value, then we can apply the probabilistic operations to find the mean color, variance or any moment of the pixel brightness values.

  

  

There are other distributions frequently used in image processing, such as **[Poisson](https://en.wikipedia.org/wiki/Poisson_distribution), [Binomial](https://en.wikipedia.org/wiki/Binomial_distribution), [Bernoulli](https://en.wikipedia.org/wiki/Bernoulli_distribution), [Reighley](https://en.wikipedia.org/wiki/Rayleigh_distribution) density functions**.

  

**Ex:** Consider a single image, $f(x,y)$ . We can assume that the pixels in an image are generated by a random source and define each color $r_k$ in this image, as a discrete random variable. First, we calculate the rough probability density function, defined as **normalized color histogram**;

  
$$p(r_k) = \frac{n_k}{n} = Prob(\text{a pixel has color } r_k),$$

where $n_k$ is the number of color with brightness value $r_k$ and $n = N \times M$ is the total number of pixels for an image of size $N \times M.$

Again, we can use probabilistic operations to find the **mean color** or **variance of the colors** or any **moment** of a given image, $f(x,y)$.



## 6. Geometric Transformation: Affine Transforms

[Affine transformation](https://en.wikipedia.org/wiki/Affine_transformation) is a [linear transformation](https://en.wikipedia.org/wiki/Linear_map) that preserves [lines](https://en.wikipedia.org/wiki/Line_(geometry)), points and [parallelism](https://en.wikipedia.org/wiki/Parallelism_(geometry)), but not necessarily [distances](https://en.wikipedia.org/wiki/Euclidean_distance) and [angles](https://en.wikipedia.org/wiki/Angle) among the pixels of a given image. Basically, it **stretches, squishies, scales, translates and rotates** an object in an image by changing the location of a pixel at $(v,w)$ to some other location $(x,y)$ .


Mathematically speaking, an affine transform $T$ is a linear mapping of a pixel at coordinate $(v,w)$ to another coordinate $(x,y)$ as follows;

$$(x,y) = T { (v,w)}.$$

In matrix form, affine transform is defined as,

$$\begin{bmatrix} x & y & 1 \end{bmatrix} = \begin{bmatrix} v & w & 1 \end{bmatrix}T ,$$

where $T$ is called $3\times 3$ affine matrix.

Design of the affine matrix is very crucial and depends on the problem domain.

An image is scaled by the following affine matrix:

$$T= \begin{bmatrix} c_x & 0 & 0 \\ 0 & c_y & 0 \\ 0 &0&1 \end{bmatrix}$$

The values of $c_x$ and $c_y$ determine the degree of scale in $x$ and $y$ directions.

An image is rotated by the following affine matrix:

$$T = \begin{bmatrix} cos\theta & sin\theta & 0 \\ -sin\theta & cos\theta & 0 \\ 0 &0&1 \end{bmatrix}$$

The angle $\theta$ defines the degree of rotation.

An image is translated by the following affine matrix:

$$T = \begin{bmatrix} 1 & 0 & 0 \\ 0 & 1 & 0 \\ t_x & t_y &1 \end{bmatrix}.$$

The amount of translation is determined by $t_x$ and $t_y$, in $x$ and $y$ directions, respectively.

**Fact**: When we apply an affine transform to an image, some of the pixels get blank or some of the pixels are mapped outside the image grid. These problems should be carefully handled. For example, we may need interpolation to fill up the color of a moved pixel. In case of the rotation straight lines become zigzagged. In order to compensate for this effect, we may need some smoothing techniques after rotation.

For example, consider a combined affine transform shown in Figure 26, where an original image is rotated and zoomed. Note that the shape of the image is no longer rectangular, but a diamond shape.

## Normalization of the Intensity Values of an Image

Most of the operations mentioned above yields an output image with non integer brightness values. Thus, when we feed digital images as the operands of an operator, there is no guarantee that we obtain a digital image at the output.  Additionally, if we work with [0-(L-1)] color range, the output image may exceed the color range. We may end up $g(x,y)$ larger than $2^k = L$ gray values.


In order to obtain images in the desired range of integer brightness values, we need to make normalization. There are a variety of techniques for normalization. One popular way is to shift and scale the color range as follows:

$$\text{Shift:}\ g_m(x,y)=g(x,y) - min(g(x,y)),$$

$$\text{Scale:} \ g_s(x,y)=round[ (L-1) \frac{g_m}{max(g_m)} ].$$

**Exercise**: Given two 2 bits/pixel digital images, as follows,

$$f(x,y)= \begin{bmatrix}0& 0 \\1 & 2\end{bmatrix}$$

and

$$g(x,y)= \begin{bmatrix}1& 1 \\2 & 3\end{bmatrix},$$

a) Find the normalized output for $f(x,y) \div g(x,y).$

b) Find the normalized output for $g(x,y) \times f(x,y).$

c) Find the normalized output for $g(x,y) + f(x,y).$

The output images should be represented by 2 bits/pixel.

**Answer:**
a) Standard division:

$f(x,y) \div g(x,y) = \begin{bmatrix}0& 0 \\1 & 2\end{bmatrix}\div \begin{bmatrix}1& 1 \\2 & 3\end{bmatrix} = \begin{bmatrix}0& 0 \\1/2 & 2/3\end{bmatrix}.$

Normalized  division with 2 bits/pixel:

$Norm(f(x,y) \div g(x,y)) = \begin{bmatrix}0& 0 \\1 & 3\end{bmatrix}.$



b) Standard  multiplication :

 $f(x,y) \times g(x,y) = \begin{bmatrix}0& 0 \\1 & 2\end{bmatrix}\times \begin{bmatrix}1& 1 \\2 & 3\end{bmatrix} = \begin{bmatrix}0& 0 \\2 & 6\end{bmatrix}.$
 
 Normalized multiplication with 2 bits/pixel:

 $Norm(f(x,y) \times g(x,y)) = \begin{bmatrix}0& 0 \\1 & 3\end{bmatrix}.$

 c) Standard  addition :

 $f(x,y) + g(x,y) = \begin{bmatrix}0& 0 \\1 & 2\end{bmatrix} + \begin{bmatrix}1& 1 \\2 & 3\end{bmatrix} = \begin{bmatrix}1& 1 \\3 & 5\end{bmatrix}.$
 
 Normalized addition with 2 bits/pixel:

 $Norm(f(x,y) + g(x,y)) = \begin{bmatrix}0& 0 \\1 & 3\end{bmatrix}.$


  
We can perform a series of operations to images without bothering the integer arithmetic. However, in order to display the output image, we need to normalize the non integer entries with the allowable bit allocation. Thus, normalization is a crucial post processing step prior to display or store images.
