# Data and Information

Although the concept of data and information are not the same, they are strongly related to each other.  

In everyday life, when we say “data”, we mean a collection of observations. On the other hand, when we say “Information”, we mean the way we understand these observations in a context.  In other words, data can be considered as  unorganized  observations, whereas the  information can be considered as structured or organized data. 

Loosely speaking, data can be defined as means of conveying information.

Both data and information have rigorous definitions in Computer Science. Let's exemplify the relation between data and information  by the following historical development of algebraic equations.

Recall the great mathematician  [Al Khwarizmi](https://en.wikipedia.org/wiki/Muhammad_ibn_Musa_al-Khwarizmi).  At his time  of the 10th century,  mathematical representation of equations were not known. He expressed an equation as follows:

“Find a number, such that, when you add its square to twice the number itself, the result is 34. “

Years later, in 17 century [Rene Descartes](https://en.wikipedia.org/wiki/Ren%C3%A9_Descartes) expressed the same information as,

$$x^2 +2x =34.$$

Al Khwarizmi and Rene Descartes expressed the same information by using different data representation. Both representations convey the exact same information.

In the above example, information is expressed by using two different  data structures. While Al-Harizmi used  the letters and words to represent a mathematical information,  Descartes introduced a new data structure of  symbols to represent the same information, which we call equation, today.

In the following, we provide the basic definitions for representing the hierarchy of  information content of an image. 

**Definition 1: Codes** are defined as a system of symbols to represent the body of information.

Codes are the lowest level of the representation of information. Examples of codes can be  letters,  natural numbers or bits.
 
**Definition 2: Codewords** are defined as a sequence of symbols to represent data.

Codewords are generated by combining the codes, as strings. Thus, they define the second level of the hierarchy.

**Definition 3: Code length** is defined as the length of the codes in a codeword.

An example of codewords  can be expression of gray levels of an image, as integers, in $[0, 255]$ in base $10,$ of natural number system.  In this example, the codewords have variable length for each codeword,  changing in $[1, 3]$.  We can also express the gray levels  of an image, as integers  in $[0, 255]$ in base 2 of binary number system, such as,  $00000000, 00000001,…$ etc. In this case, the code length is fixed to $8$. 

**Definition 4: Data** is defined as the collection of the codewords to represent the body of information.

The term data refers to multiple set of codewords to represent information. A single codeword of data is called **datum**.

Data can be considered as the third level of abstraction to represent information. Data of an image can be the collection of pixels, associated with coordinates and color values. 

**Definition 5: Alphabet or Codebook** is the set of codewords to represent the data.

The codewords of an Alphabet is used to represent the  body of information. For example, an alphabet of an image may consist of the code words, which represent the colors of an image in the  palette.

Codes, codewords and data can be considered as  lower levels of abstraction to represent information. 

The concept of information is much more complicated  and abstract then the concept of data. Literally, it means the power to inform.  The word **inform** comes from Latin word **informatio**, which means **to give form to mind**. Thus, liteally, a collection  of data, which shapes our mind can be considered as information.


Information  is mathematically defined by [Claude Shannon](https://en.wikipedia.org/wiki/Claude_Shannon) in 1948, in a pioneering  paper titled ["A Mathematical Theory of Communication"](https://en.wikipedia.org/wiki/A_Mathematical_Theory_of_Communication).  He developed [Information Theory](https://en.wikipedia.org/wiki/Information_theory) to formalize the quantification, storage and communication of  information, using probability theory and statistics.  The starting point of Shannon was  to model the information as the **uncertainty to be resolved in a set of data.** Therefore, if one can formally define the uncertainty, this allows us to define information, which removes  uncertainty.

# Measuring Information 

Around the 1940s,  Shannon was working as an engineer in a telephone company, called Bell Labs. Obviously, at this time there were no wireless networks:) His goal was to measure the capacity of telephone cables according to the household demands of making the phone calls.  Shannon was seeking a measure of information to be carried on the telecommunication network  to estimate the capacity of cables. He asked the following critical question to himself:

**Motivating Question:** How few data is needed to store or communicate the same amount of information?

He attempted to find an answer to this question by using the concept of **uncertainty** in Probability Theory. This is how Information Theory was born from the heart of Probability Theory. 

Loosely speaking, uncertainty is defined as the **limited knowledge** about the outcome of a probabilistic event.

In probability theory, a probabilistic event, $E$, can be defined by a two-tuple, $\{x, p(x)\},$ where  $x$ is a random variable, which maps all possible outcomes of a probabilistic event $E$ to a set of real numbers. In other words,

$$x: E \rightarrow R,$$

where $R$ is the space of real numbers. If we have a discrete random variable, then, $R$ is replaced by the space of integer numbers, $I$.

The probability density function, $p(x)$, shows the relative frequency of all possible outcomes  that an event $E$ generates.

Figure 2: Probability density function, $p(x)$ of a Gaussian random variable with low uncertainty (red) and low (blue)  uncertainty.</figcaption></p></figure>

Intuitively, uncertainty of a random variable, $x$, is somewhat related to its probability density function, $p(x)$. 

Consider a discrete random variable, $x$. If the probability density function is very sharp, with a single nonzero value for a particular outcome of $x = \mu$, given below,


$$p(x ) = \begin{cases}1\quad for \quad  x = \mu 
\\\ 0 \quad o.w.\end{cases}.$$ 

then, we certainly know that the outcome $\mu$ of an event, $E$, will definitely occur. Therefore, this outcome is **certain** and when it occurs it is **not a surprise** for us.  As  the probability density function becomes flatter, the uncertainty of this event increases.  The **highest uncertainty**  can be obtained for a **uniform distribution**, when we get all the outcomes equally likely.

In the above intuitive argument, information can be considered as the degree of surprise of a random event, $E$, represented by the outcome of the random variable $x$. For a deterministic event, when an outcome has $p(x =\mu)=1$, there is no uncertainty for this outcome and we do not get surprised at all. **Thus, information can, also, be considered as the degree of surprise.** This  reveals that it is possible to formally define information as the **degree of uncertainty to be resolved,** by  a function of the probability density function, $p(x)$. 

In order to define this function, Shannon had three basic criteria: A measure of information is to be
-	intuitively correct, 
-	a positive quantity and
-	a [monotonically nondecreasing function](https://en.wikipedia.org/wiki/Monotonic_function) of p(x).

The above three criteria can be satisfied by a well-behaved function, called **logarithm**.  However, since the probability measure,  $0\le p(x)\le1$, the logarithm of it, $\log p(x)$,  is always negative, which violates the second criteria. Then, it would be wise to take the negative of the logarithm of the probability measure. Luckily,  logarithm is a monotonically nondecreasing  function of its argument, as shown in Figure 3. 

<figure><p align="center"><img src="https://raw.githubusercontent.com/hazalmogultay/466Images/master/W10/log_graph.png" height=300px><figcaption>Figure 3: Logarithm of the probability density function, $p(x).$ </figcaption></p></figure>

Thus, we can define a measure of information as follow:

**Definition: Self Information** of a probabilistic event, $E$, represented by the random variable  $x$ and its associated probability density function, $p(x)$, is defined as,

$$I(x)=\log\frac{1}{p(x)}=-\log(p(x)).$$

The unit of self information is defined with respect to the base of the logarithm. If it is a base-2 logarithm, the unit is called binary units, in short **bits.** If the base is $m$ then, it is called **m-ary units**. If the base of the logarithm is $e$, then it is called **nats**. 

**Motivating Question:**  Is the above definition of self information intuitively correct?

Figure 4 indicates a typical plot of self information vs. probability density function $p(x)$. Notice that $I(x)$ is inversely proportional to  $p(x)$.

Figure 4: A typical plot of self information, $I(x) = log[1/p(x)].$

Lets analyze the self information at the extreme cases:

**Case 1:** Suppose that,

$$p(x= \mu) \rightarrow 1 \implies I(x= \mu) \rightarrow  0.$$

In this case, the uncertainty decreases and eventually the event becomes deterministic  and   there is no uncertainty about this event. When it happens, there is no surprise. Therefore, no information is needed to resolve this event. 

**Case 2:** Suppose that,

$$p(x =\mu) \rightarrow  0 \implies I(x=\mu) \rightarrow \infty.$$

This means that the event, $E$, never  happens at $x= \mu$. If we observe that this event  happens, then  there is an infinite amount of  surprise, since
$$I(x= \mu) \rightarrow \infty.$$ 

Therefore, we require an infinite amount of information to resolve this event.

Thus, high surprises and high uncertainties bring high information value, which is intuitively meaningful.

Let us measure the information content of a very simple event, in the following example.

**Example:** Recall toss  the coin  experiment in probability theory, where the event, $E$, is the tossing a coin with two outcomes, **Tail** and **Head**. Suppose the random variable x=1 represents **Tail** and x=0 represents **Head** outcomes. 

**Case 1:** If the coin is unbiased, then,  $P(x=0) = P(x=1) = \frac{1}{2}$, For this case, each outcome has a self information as

$$I(x=1) = I(x=0) = 1\quad \text bit.$$


Thus, one bit of information is required to represent each outcome of the toss the coin event.  
In other words,  1-bit is the amount of information required, when one of two equally likely events occur.


**Case 2:** If the coin is biased such that we always get **Tail**, then,
$P(x=1) = 1, \quad \text {and}\quad  P(x=0) = 0.$ 

For this case,  the self information for observing **Tail** and **Head** is, 

$$I(x=1) = 0 \quad \text {bit}  \quad \text and \quad I(x=0) = \infty\quad \text {bit,}$$
respectively. In other words, when we observe **Tail**, this observation does not carry any information. However, when we observe **Head**, this observation carries an infinite amount of information.




# Entropy: Average Information

Recall that, an $N\times M$ image with k-bits per pixel requires  a space allocation of  $N\times M\times k$.  We can easily reduce the size and bit allocation of this image by sampling and quantization techniques. However, as we decrease the space allocation, the quality of image starts to degrade, due to information loss. Therefore, we need to somehow measure the degree of information loss, to design an image compression algorithm. Therefore, our goal is to develop an image compression algorithm to store or transmit an image with a minimum possible space allocation without decreasing its information content. 

**Motivating Question:** What is the minimum space for storing an image with the same amount of information? 

Although the definition of self information was a good starting point, it falls short in measuring the amount of information of an image. It only measures the information content of a single outcome of an event, which corresponds to the codeword of  a single pixel color.  On the other hand, we need to measure the information content of the entire  image, which corresponds to the collection of codewords, representing all the pixel colors. We need  a measure of information, which sets  an upper limit for reducing the amount of data to represent the same amount of  information.


A very crucial  concept of Information Theory, which can be used to measure the  amount of information of an image is called **Entropy**.  

Entropy is originally defined in Thermodynamics. Roughly speaking, it is a  measure of disorder or uncertainty,  in a closed system.  Shannon adopted this concept to communication systems. Later, it became one of the core and controversial concepts in many fields of science and technology.

Let us follow the path of Shannon and consider an information channel, which is characterized by 
-	a source, which generates information, 
-	a channel where the information is conveyed or stored and 
-	a sink, where the information is received and consumed, 

as shown in Figure 5.

Figure 5: Representation of an information channel by source, channel  and sink alphabets with their associated probabilities,  $(A,Z)$ and $(B,V)$, respectively and channel transmission matrix, $Q$.</figcaption></p></figure>
  
In an information channel, the source is represented by a set of codewords from a finite alphabet,  called the **source alphabet,**

$$A=\{ a_1, \ a_2, \ \ldots ,a_J \}.$$ 


The probability of occurrence of each codeword in the alphabet  is defined by the set,  

$$Z=\{ P(a_1), \ P(a_2), \ldots P(a_J),\},$$

 where $\sum^{J}_{i=1} P(a_i) =1$. 
 

Thus, the source of an information channel  is represented by a two tuple,  $(A, Z)$, with an alphabet $A$ and the probability set of codewords, $Z$

Suppose that we want to generate a codeword, $a_i$, from the alphabet $A$, at the source. Then, as defined in the previous section, the self information for generating this codeword is

$$I(a_i) = -\log P(a_i).$$


**Motivating Question:**  Suppose now that we need to generate a  sequence of codewords, $\{a_i\}$, from the alphabet, $A$.  What is the average information required to send each codeword?

The answer to the above question can be given by defining a new concept, which is called **Entropy.**

**Definition:** Given a sequence of symbols drawn from a finite alphabet $A=\{ a_1 \ a_2 \ \ldots a_J \},$ 
 with their probabilities, $Z=\{ P(a_1) \ P(a_2) \ldots P(a_J)\}$, **Entropy** is defined as the average information generated at the source output. Formally, Entropy is defined as,

$$\boxed{H(a_i) = \sum^J_{i=1} P(a_i) I(a_i) = -\sum^J_{i=1}  P(a_i) \log P(a_i)  \quad \text{unit/codeword}}.$$

The unit of entropy depends on the base of the logarithm. As in the most popular case, if it is a base-2 logarithm the unit is **bit.** 


Similarly, we can represent the sink of an information channel by a set of codewords from a finite set of alphabet,  called the sink alphabet, 

$$B=\{ b_1, \ b_2 ,\ \ldots ,b_J \}.$$ 


The probability of occurrence of each codeword in the sink alphabet  is defined by the set,

$$V=\{ P(b_1), \ P(b_2), \ldots ,P(b_J)\},$$

 where $\sum^{J}_{i=1} P(b_i) =1$. 

Thus, the sink of an information channel  is represented by two tuples,  $(B,V)$, with an alphabet $B$ and the probability set of each codeword, $V$.

Finally, we can represent the channel, by a channel transmission matrix $Q$, defined as,

$$Q=\{ P(b_k/a_i)\},$$

where each entry of this matrix is defined by, 

$$P(b_k) = \sum_{i=1}^JP(b_k/a_i)P(a_i).$$



Therefore, entropy can be considered as the average information per codeword, measured in collection of data, generated by a probabilistic source. It can, also, be considered as a **measure of uncertainty** of an event generated by a probabilistic source. 


The above definition of entropy, can be generalized to any event generated by a probabilistic source. It has a wide range of impact in many applications.

## Coding Theorem

Coding theorem, suggested by Shannon, uses the concept of entropy to establish an upper bound for compressing the data without loosing information . 

Suppose that we are given a  probabilistic source, represented by and alphabet and probability distribution function. Suppose also that we allocate codelength of $l(r_k) $ bits to represent each codeword, $r_k$,  in the alphabet. The theorem estimates the average codelength, allocated to each symbol in the alphabet.  It also estimates the entropy (average amount of information per codeword) of the source  Then, it compares the average codelength to the entropy of the source and shows that it is impossible to compress the average codelength of $l(r_k)$  bits  less than the estimated  entropy value of the information source. 


Let us apply the coding theorem to  images by comparing, 
-	the entropy of an image, which is the average amount of information contained in a color, 
-	the average amount of  allocated **data** to store or transmit a  color. 

The average amount of **information** contained in a color can be measured by entropy as,

$$ H= -\sum p(r_k) log(p(r_k)) \quad \text {bits/pixel.}$$

The average amount of  allocated **data** to store or transmit a color can be measured by average code length  to represent this color. Suppose that, we allocate a codelength of $l(r_k)$ bits to store a  pixel color, $r_k$. Then, the average amount of code length to represent all the colors, $r_k, \quad \forall  k= 0,1,...(L-1)$, in the palette , can be defined as,

$$ L_{av,L} = \sum_{k=0}^{L-1}p(r_k) l(r_k) \quad \text {bits/pixel,}$$

where $l(r_k)$ is the allocated  code length for color, $r_k$ and $L$ is the total number of colors. Comparison of the above equations show that while entropy is the average value of  information, the average code length is the  average value of the allocated code length of pixel colors.

 **Coding theorem**,  gives a lower bound for the average number of codelengths to  store or transmit images, as follows;
 
$$\lim_{L\rightarrow \infty} [\frac{ L_{av,L}}{L}] \rightarrow H.$$ 

The above theorem states that the lower bound we can ever get for the average number of codelengths in a compressed image approaches entropy, as we increase the number of colors, $L$.


## Redundancy and Compression Ratio

Coding theorem suggests that, when we design a compression algorithm, our goal is to reach the lower bound for the average code length of a pixel in the image, which is measured by entropy. There are two related quantities to be measured to develop a compression algorithm, namely,
1. Redundancy, which quantifies the irrelevant information, embedded in the original image.
2. Compression ratio, which quantifies the degree of compression, obtained at the output of the compression algorithm.

Let us provide the definitions of the redundancy and compression ratio.

**Definition: Redundancy** is defined as the difference between the average information (entropy)  and average allocated code length, as follows;

$$R=  L_{av,L} - H.$$


Note that, when the  allocated code length of a color is the same as the self information of $r_k$,  $$l(r_k) = I(r_k), \quad \forall k, $$
then, the redundancy becomes, $R =0.$

When we design  a compression algorithm, the degree of compression  is defined as follows.

**Definition: Compression ratio** is defined as

$$C_R = \frac{n_1}{n_2}= \frac{\text{Number of bits in a given image}}{\text{Number of bits in the compressed image}}.$$

Compression ratio depends on the redundancy embedded in image content. If an image has large smooth areas, the compression ratio gets large. On the other hand, if there are a lot of color variations in small image pieces then, the ratio gets smaller.  The relationship between the compression ratio and redundancy is defined by the **relevant data redundancy**, defined  as follows,

$$R_D = 1-\frac{1}{C_R}.$$

Note that, when 

$$\\n_1=n_2\implies R_D=0 : \text{no redundancy} \\ n_1>>n_2\implies R_D=01: \text{high redundancy}$$

Image compression algorithms try to minimize the redundancy or relevant data redundancy to increase the compression ratio.

##  Measuring Information to Compress Images 

Coding theorem, mentioned above, can be used to estimate the average amount of codewords, $L_{av}$ to be allocated  to store the pixel colors of an image without losing   information content of the image. 

Let us study how we can measure the minimum amount of data to store or transmit an image  by the following example. 

**Example:** Consider the following image:

$$f = \begin{bmatrix} 0 & 0 & 0 & 32 & 64& 128 & 128 & 128\\ 0 & 0 & 0 & 32 & 64& 128 & 128& 128\\ 0 & 0 & 0 & 32 & 64& 128 & 128& 128\\ 0 & 0 & 0 & 32 & 64& 128 & 128& 128\\ \end{bmatrix}.$$

Suppose that our goal is to compress this image piece with the least amount of storage space. What is the minimum required storage space?

**Answer:** The answer depends on  the way we measure the amount of information embedded in this image.

**Case 1:** If we  know nothing about the entropy and self information of the color content  of the image, we just inspect the  color range in $[0, 128]$ of the palette of this image. This rough analysis shows that we  need $L_{av} = 8$ bits per pixel to code the colors in the palate, $r_i= 0, 32, 64, 128$. Thus,  we need a storage space of  4x8x8 = 512 bits for this image.

The above coding method, implicitly,  assumes that occurrence of each color is equally likely. In other words, color distribution is uniform distribution, i.e., 

$$p(r_i) = \frac{1}{256}.$$ 

Assuming uniform distribution, if we compute the entropy of this image, we find that,

$$H =\sum_{i=0} ^{255}\frac{1}{256} log \frac{1}{256} = 8 \quad \text {bits/pixel.}$$ 

**Case 2:** Assume that the colors, $r_i$ of the  image is generated by Shannon’s probabilistic source. In order to model the source, we define a source alphabet of colors and its corresponding probabilities of occurrence of each color as follows;

$$A= \{r_i\} \quad \text and \quad Z=\{p(r_i)\},$$
respectively.

<figure><p align="center"><img src="https://raw.githubusercontent.com/hazalmogultay/466Images/master/W10/image28.jpg" height=300px><figcaption>

Figure 8: Normalized histogram, where the colors in the palette correspond to the codewords of Source alphabet, $A$. The set, $Z$, consists of the relative frequencies, $p(r_k)$,  of each color, in the palette of the above image. </figcaption></p></figure>

The set $Z$, consists of the relative frequency of occurrence for each color and can be easily approximated by the  normalized histogram $p(r_k)$, as follows;

 $$p(r_k = 0) = p(r_k = 128) = \frac{3}{8} \quad \text and \quad p(r_k = 32) = p(r_k = 64) = \frac{1}{8},$$ 

as shown in Figure 8.

Let us now  compute the entropy of this image, 
$$H=-(2\frac{3}{8}log\frac{3}{8} + 2\frac{1}{8}log\frac{1}{8} ) = 1.81 \text{bits/pixel}.$$ 

The above entropy measure shows that instead of $L_{av} = 8$  bits/pixel, we need  $H = 1.81$ bits/pixel to store this image. Thus, according to Shannon’s Information Theory we need a total of 8x4x1.81 = 58 bits to store this image.  The difference between $L_{av}$ and   $H$  is the redundancy, $$R = L_{av} - H$$  of the coding schema.

 Thus, we can develop an algorithm to compress this image to 58 bits.

**Note:** Recall that defining a probabilistic source requires defining the two-tuple, an alphabet and the probability distribution of the codewords in the alphabet. Thus,  definition of the entropy depends on how we model the image as  a probabilistic source.  


## Second Order Entropy

In the context of this course, we suppose that images are  generated by a probabilistic source. We can model the probabilistic source by using more than one random variable.  For example, we can define joint probabilities of two random variables, which measures the relative frequency of occurrence of colors, $(r_i, r_j)$ of two neighboring pixels. This model defines a joint probability distribution $p(r_i, r_j)$ and the corresponding entropy defined below is called the **second order entropy**:

$$\boxed{H(r_i, r_j) = -\sum_i \sum_j p(r_i,r_j) \log p(r_i, r_j) \quad \text{bits/pixel}}.$$

It is possible to define higher order entropies by modeling the probabilistic relations among the pixels in a given neighborhood system, where we assume that the probabilistic source generates the relationships among multiple pixels in a predefined neighborhood.  Figure 9 shows the possible pairing of pixels to measure the relative frequency of occurrence in pairs. Selection of one of these pairs depends on the application. Or we may compute the entropy for all cases, to measure the interpixel redundancy.



<figure><p align="center"><img src="https://raw.githubusercontent.com/hazalmogultay/466Images/master/W10/L10.png" height=150px><figcaption>Figure 9: Possible pixel pairs can be used to define second order entropy. This representation enables us to estimate the joint probability of occurrence of pixel pairs. </figcaption></p></figure>

Higher order entropies measure the information content of the image to reduce interpixel redundancy. In order to get a glimpse of second order entropy, let us consider the image given above. As a third case let us compute the second order entropy, in the following example.

**Example: Consider the following image:

$$f = \begin{bmatrix} 0 & 0 & 0 & 32 & 64& 128 & 128 & 128\\ 0 & 0 & 0 & 32 & 64& 128 & 128& 128\\ 0 & 0 & 0 & 32 & 64& 128 & 128& 128\\ 0 & 0 & 0 & 32 & 64& 128 & 128& 128\\ \end{bmatrix}.$$

Suppose that our goal is to compress this image piece with the least amount of storage space. What is the minimum required storage space?

**Answer:** Assume that the image has a taurus structure (leftmost column is attached to the rightmost column), we can estimate joint probability of pixel pairs $p(r_i, r_j)$ by just counting the number of occurrences of each color pair of  two horizontally neighboring pixels, as follows;

$$p(r_i=0,r_j=0) = p(r_i=128,r_j=128) = \frac{1}{4}$$ 
$$p(r_i = 0,r_j = 32) = p(r_i = 32,r_j = 64) = p(r_i = 64, r_j =128) = p(r_i = 128, r_j =0) = \frac{1}{8}.$$ 

The corresponding second order entropy is found as, 

$$H(r_i, r_j) = -(2\frac{1}{4}log\frac{1}{4} + 4\frac{1}{8}log\frac{1}{8})=1.25 \quad \text {bits/pixel}.
$$   


 The second order entropy measures the information content of the image considering the two neighboring pixel pairs as  random variables. Since the same horizontal patterns are repeated in the above image, we could reduce the information content of the image  by using the statistics of pixel pairs, $p(r_i,r_j)$. This model allows us to reduce the average code length allocation, which requires $L_{av} = 1.25$ bits per pixel pair to code our image.

We could, also, define and estimate the second order entropies using vertical or diagonal pixel pairs, as shown in Figure 9. Also, it is possible to define higher order entropies by considering larger neighborhood systems. 

In the context of this course, we shall suffice to use first order entropy and we shall use the normalized color histogram to estimate it.  



# Image Compression Methods

Measuring the entropy of an image provides a lower limit for the storage allocation of an image. However, it does not tell us about the methodology to compress an image to store it with the measures. Image compression methods aim to reach the lower bound of the storage allocation estimated by the entropy by reducing the redundancies embedded in the image. 

In order to develop compression algorithms, first we need to define the type of the compression methods. There are two major approaches for compression:

**1. Lossy compression**, in which we can tolerate a certain amount of information loss.

**2. Lossless compression** , in which we keep the whole information, such that when we decompress it, we obtain the exact same image as the original one.

Suppose that we have an image, $f(x,y)$ and we  compress it in the source. Suppose, also, that we decompress this image at the sink and obtain the recovered image, $\hat{f}(x,y)$, as shown in Figure 9. If the compression is lossy, then the error between the original image and decompressed image is,

$$e(x,y) = \hat{f}(x,y)- f(x,y) \neq 0 ,\quad  \forall (x,y).$$
If the compression is lossless, then, the error is,

$$e(x,y) = \hat{f}(x,y)- f(x,y) = 0 ,\quad  \forall (x,y).$$



The type of the compression method depends on the application domain. For example, if we store an image for human consumption, we may tolerate losing some information, which is beyond the perception limits  of the human visual system. On the other hand, if our aim is to design a high speed target detection algorithm, such as hitting a moving target with a fast speed,  every bit of information is critical to achieve  high performance. Therefore,  in order to satisfy the requirements of this type of an  algorithm, we develop lossless compression algorithms.

# An Image Compression  Model as an Information Channel

Image compression methods can be modeled as an information channel, which consists of  two major modules, 

**1. Encoder,** which compresses the image in the source, to reduce the image size.

**2. Decoder,** which decompresses the image, in the sink, to recover the original image (see Figure 11).

In the source, there are two  encoders. The first one, called, **source encoder,**  reduces all types of  redundancies in the image. The second one, called, **channel encoder,** codes the image to make it robust to channel distortions. Channel encoders add extra bits to the source-encoded image to detect the transmission interferences and correct them. 

In the sink, decoders reverse the encoding processes to recover the original image, as shown in Figure 11. In this course, we shall omit the channel encoding and decoding algorithms, since they require background on network communication methods. We just  focus on the **source encoding** methods to reduce the redundancies.

## 1. Encoding in the Source

Source encoder consists of algorithms, which reduces the redundancies in three major modules:

**1. Mapper,**  reduces the interpixel redundancy.
**2. Quantizer,**  reduces the phsycovisual redundancy.
**3. Symbol Encoder,**  reduces the coding redundancy.

The algorithms in the encoder, basically,  reduce the redundancies mostly by using some heuristics, which involve optimization techniques. Once we measure  several order entropies of an image, we get a rough idea about the information content, thus, we can estimate the amount of redundancy embedded in the image. Then, we develop algorithms to reduce  three different types of redundancies, in the mapper, quantizer and symbol decoder to compress the images with the “optimal average code length** and  compression ratio. 

##  2.  Decoding  in the Sink

Once the image is compressed in the source encoder, it is either stored in a local storage device or further encoded for transmission. In order to consume the image in the sink end, it should be decompressed by reversing the  operations of the source end. 

In the sink, the image is, first, decoded by the channel decoder. Then, it is fed to source decoder to recover the original image in three modules, as shown in Figure 13,

**2. Dequantizer,** recovers the original colors

If the encoder algorithms use lossless compression methods, then the decoders simply implement the inverse of these algorithms.

If the encoder algorithms use lossy compression methods, then, decoders should use some approximation techniques to recover the original image as much as possible. # Algorithms for Reducing the Redundancies, in Source Encoder

Recall that we defined three types of redundancies embedded in an image:

1. Interpixel Redundancy
2. Psycho-visual Redundancy
3. Coding Redundancy


In the following subsections, we suggest simple algorithms to reduce the above types of redundancies, in the mapper, quantizer and symbol decoder.


## 1. Reducing the Interpixel Redundancy, in the Mapper

Images represented in spatial domain consist of a lot of repetition in pixel color values or patterns in large neighborhoods. In video recordings many frames are the repetition of a reference frame with minor changes. These types of spatial and temporal repetitions are called interpixel redundancy.  

Interpixel redundancies can be categorized under three groups:

1. Spatial redundancy,  where the same color is repeated in large image patches.
2. Geometric redundancy, where the same pattern or shape is repeated to form a texture.
3. Interframe redundancy, where the video frames are repeated from a reference frame.

Unfortunately, inter pixel redundancy cannot be measured by first order entropy. Because first order entropy does not take into account the probabilistic relations among the pixels in a neighborhood system. Images with the same  normalized histograms may carry different  types of interpixel redundancies. A very good example is shown in Figure 14.  The images in this figure have exactly the same histograms.  Therefore, the first order entropies are the same.  However, a visual inspection shows that the image on the left is more organized than the one on the left. 

  <figure><p align="center"><img src="https://raw.githubusercontent.com/hazalmogultay/466Images/master/W10/uniform.png" height=300px><figcaption>Figure 14: These images have the same histogram. However, the one on the left is more ordered, thus, it has more interpixel redundancy.</figcaption></p></figure>

In order to measure the information content of the relationships among the pixels, we  need higher order entropies, defined below;

$$H(r_i, r_j,..r_l) = -\sum\sum… \sum P(r_i,r_j,...,r_l) \log P(r_i, r_j,...,r_l) \quad \text{bits/pixel},$$

High order entropies  measure the information content among multiple pixels. Thus, it provides a measure  for  interpixel redundancy. Determining the order of the entropy may not be easy in many practical problems. Thus, we need other measures for finding a lower bound for average code length for an image.

A useful function for  measuring interpixel redundancy is the **auto-correlation function** defined below;

$$R(x,y) = \frac{1}{NM} \sum_{\forall k} \sum_{\forall l} f(k,l)f(x+k, y+l), \forall x,y.$$

Note that autocorrelation operation measures the correlation of the two dimensional image function  with itself. It carries the properties of convolution, except the negation operation. Figure 15 shows an image and its autocorrelation function. By inspecting the autocorrelation function, one can get an opinion about the interpixel redundancy embedded in this image. The more the autocorrelation function (ACF) gets flatter in an interval, the more  interpixel redundancy is embedded in the image. For example, if we had another circle in the image the ACF would have two bumps instead of one. 

Since the autocorrelation operation translates the image into itself,  we need to augment the image by using a padding schema, for example we can assume that the image is padded by a taurus structure.


  <figure><p align="center"><img src="https://raw.githubusercontent.com/hazalmogultay/466Images/master/W10/acf.png" height=250px><figcaption>Figure 15: a) A simple image of red circle, b) its one-dimensional cross section, c) the autocorrelation of the cross section and d) the autocorrelation of the two dimensional image.  </figcaption></p></figure>

**Motivating Question:** How do we reduce interpixel redundancy embedded in an image?

There is a wide range of techniques to reduce the interpixel redundancy, some of which we already have seen. 


### Image Transforms

Recall  that images can be represented in transform domain, as a linear combination of a set of predefined basis functions, as follows,

$$f(x,y)  = \sum_{x=0}^{N-1}\sum_{y=0}^{M-1} T(u,v) h(x,u,y,v),$$

where $h(x,u,y,v)$ is the set of basis functions, which span the transform domain and the entries  $T(u,v), \forall (u,v),$ of the transform matrix, $T$, are the coordinates of the image in the transform domain. Depending on the selection of the basis functions, $h(x,u,y,v),$ the transformed image, $T$, can be represented in one of the **Fourier, Cosine, Hadamard or Wavelet domains.** We have already seen that transform domain images are very compact. This representation allows us to reduce the size of the image significantly. 

  <figure><p align="center"><img src="https://raw.githubusercontent.com/hazalmogultay/466Images/master/W10/13.png" height=250px><figcaption>Figure 16: Cosine Transform of Lena image. The transformed image has a lot of zero values, which can be discarded for compression. </figcaption></p></figure>

Figure 16 shows the Cosine transform of the Lena image. Note that the transformed image is mostly squished to the upper left corner of the image. Thus, by keeping the nonzero values of the transformed image in the upper left corner we can compress the image. The full image can be reconstructed from its compressed transformed image. In other words, we can truncate the  limits of the summation,

$$f(x,y)  = \sum_{x=0}^{K-1}\sum_{y=0}^{L-1} T(u,v) h(x,u,y,v),$$

to remove the zero values in the transform domain, for $K\lt N$ and $L\lt M$. This truncation corresponds to ideal low pass filtering of the image, in transform domain, which results in a compressed image.


Therefore, rather than compressing an image in spatial or temporal domain, we transform it to one of the  domains, then, apply a set of compression algorithms to reduce the redundancies in the transform domain. The popular JPEG image file format first partitions the image into  $8\times 8$ blocks, then applies a series of encoding algorithms to  reduce interpixel redundancy, in the cosine domain.

Also, techniques, such as filtering, subband coding and image trees, can be used to further reduce spatial redundancy, in transform domains. 

In the following, we provide two  additional techniques, namely, **run length coding** and **bit plane slicing,** to reduce the interpixel redundancy. 



### Run-length Coding

A very simple method to reduce the interpixel redundancy is called **run length coding (RLC)**.  It is used in many compression codecs. It is, also, used to compress binary document images for facsimile machines. 

RLC reduces the size of a repeating string of symbols (i.e., runs) by generation a list, where each entry of the list consists of two bytes:

1. Code of the symbol
2. Count of the repetition of that symbol.

For example,  given a string of symbols, below:

$$S= (aaaabbbbbcccccc),$$

 run length representation of this string is,

$$RLC(S) =(a,4), (b,5), (c, 6).$$
Similarly, if we had a binary string, given below,		

$$B= (11111000000011111111),$$

its  run length representation of this string is,


$$RLC(B) =(1,5), (0,6), (1, 8).$$

The above idea of RLC of string of symbols can be applied to binary images. Consider the binary image, given in Figure 17. Rather than representing this image as an array or matrix form, we can generate a list of lists, where only the black pixels are coded by three attributes as follows;

-	row number, 
-	starting column number and 
-	ending column number.

Each black pixel is coded by three attributes: row number, starting column number and ending column number, as shown in the right hand side. </figcaption></p></figure>

For the above image, we prepare a list of lists  as follows;
$$I = (\underbrace{0}_{row}\ \underbrace{2}_{start\ column}\ \underbrace{2}_{end\ column}) (1\ 1\ 4) (2\ 1\ 5)  (3\ 1\ 6)  (4\ 1\ 6).$$

**Note:** RLC is a lossless compression algorithm.

### Bit Plane Slicing 

Consider an NxM image array with $2^k = L$ distinct colors. In this representation, each pixel has a codeword of length $k$. We can represent each bit of  image pixels by binary images. Thus, the image is decomposed into k-binary images, as shown in Figure 18.

Then, we can process each bit plane to remove the interpixel redundancy. For example, we can use run length coding for each bit plane, to reduce the redundancy. 

In most practical applications, the least significant couple of bits do not carry significant information. In this case we may discard the least significant bit planes, as long as we cannot discriminate the difference between the compressed image and the original image.

## 2. Reducing the Psycho-Visual Redundancy, in the  Quantizer

Recall the characteristics of the human visual system. We, humans,  perceive the physical world in a somehow distorted and approximated way, so that visual illusions can trick us. We, also, make mistakes while we perceive color or contrast. Our visual system can only perceive a limited resolution and color bands. We cannot discriminate colors if they are closer to each other than our perceptible limitations. Moreover, our algorithms consist of integers to represent colors. Thus, when we display an image using numbers and electronic devices, we should keep in mind that this is just an approximation of what we see.

An image compression algorithm should take into account all the characteristics and limitations  of the human visual system, so that original and decompressed images should look the same for the human visual perception.

There are three major approaches to remove the psychovisual redundancies

-	Downsampling and interpolation, which reduces the spatial resolution,
-	Vector Quantization, which groups the similar image patches and codes each of them with an index. 
-	Color quantization, which reduces the number of colors.



### Downsampling and interpolation

Recall the pyramid and tree representation of images, where we sample an image at each level and use an approximation filter to assign the color of the pixels at each level of reduced resolution. In this representation, we can eliminate the images in high resolutions, as long as we cannot discriminate the difference between the levels. We keep the level of the tree with the Just Noticeably Different (JND) image level.  In order to decompress a downsampled image we may want to interpolate the upsampled images to improve the smoothness (see Figure 20).

### Vector Quantization

An image is made of very similar patches, which cannot be discriminated against by the human visual system. 
     

 <figure><p align="center"><img src="https://raw.githubusercontent.com/hazalmogultay/466Images/master/W10/9.png" height=300px><figcaption>Figure 20: Image in the left has hundreds of 8x8 blue patches, shown in the right, which cannot be discriminated against by the human visual system. Rather than repeating the similar patches, we cluster them and code the clusters of similar patches. </figcaption></p></figure>


The similar patches  can be clustered under the same group and the patches under a group can be coded under the same codeword. There are a large variety of vector quantization methods in the literature. In the following , we provide a very simple method.  

**Algorithm**: A Simple Vector Quantization Method

Input: An image database, which consists of array representation

Output: Vector quantized image and  a codebook 

1. Divide the original  images into blocks (for example 8x8 blocks), represent each block by a vector $X_i=(X_{ii} \ \ldots \ X_{ni})$
2. Cluster similar vectors under the same category, for example using K-means clustering method. 
3. Code vectors under the same cluster by a code word to generate code-block
4. Represent the image by the code words,  in the codebook, each of which corresponds to an image patch.

> **Note:** We may take the transforms of the original images, prior to vector quantization. Then, we may run the above vector quantization algorithm in the transform domain.


## Color Quantization

Color quantization methods reduce the number colors of the palette of an image. This method reduces the code length of colors. 

There are many methods for color quantization. Most of them split the histogram into intervals of colors and select a single color to represent an interval. A practical way of partitioning the histogram into color intervals is to find the valleys and the peaks of the histogram. Then, the peak value between the two valleys can be taken as the representative color of this particular histogram. 


  <figure><p align="center"><img src="https://raw.githubusercontent.com/hazalmogultay/466Images/master/W10/quant2.png" height=250px><figcaption>Figure 21: An example of histogram of an  image. We can select the colors at the peak values $r|1, r_2$ and $r_3$ and discard the rest of the colors. Then, each interval  between the valleys is represented by the most frequent color in that interval. </figcaption></p></figure>

### A fast color quantization algorithm using a set of one dimensional color intervals by [M. Uysal and F.T. Yarman Vural](https://ieeexplore.ieee.org/abstract/document/723455)

Considering the fact that color histograms are not smooth enough,  finding the valleys and peaks may not be an easy task. Additionally, some rare colors may be crucial to characterize the image if they provide high contrast. 

A better approach for color quantization can be achieved by selecting the most frequent and most discriminative colors from the Palette of an image. M. Uysal and F.T. Yarman Vural suggested a color quantization method, which minimizes the total  error function, as defined below;

$$Error = \sum_{\forall k} \sum_{\forall j} p(r_k )\Delta^2(r_k,r_j), $$ 
iteratively, by selecting $J$ colors from $L$ colors. In the above formulation,  the distance between color $r_k$ and color $r_j$ is defined as,

$$\Delta^2(r_k,r_j) = (x_k - x_j)^2 + (y_k - y_j)^2 + (z_k - z_j)^2 , \forall r_j.$$



They suggested an iterative algorithm to efficiently select  $J$ representative color, $r_k$, out of $L$ colors for which minimizes the total error,  given below:

**Algorithm: Mutlu’s Color Quantization**
Inpun: An image with L colors
Output: An image with J colors, where $J \lt L$

1. Find the color histogram of the image.
2. Make a list from the most frequent to least frequent colors.
3. For all colors, $r_k = (x_k \ y_k \ z_k),$ measure the distances between the rest of the colors, $r_j, \forall j\neq k,$ by using the Euclidean distance, given below;

$$\Delta^2(r_k,r_j) = (x_k - x_j)^2 + (y_k - y_j)^2 + (z_k - z_j)^2 , \forall r_j.$$

4. Define color similarity matrix (CSM)  for Palette $P=\{ r_k\}^{L-1}_{k=0}$
  <figure><p align="center"><img src="https://raw.githubusercontent.com/hazalmogultay/466Images/master/W10/palette2.png" height=300px><figcaption>Figure 22: Color similarity matrix (CSM). The leftmost column has the sorted colors from the most frequent to least frequent color. Each row of the color similarity matrix shows the sorted color with respect to the Euclidean distance of the color at the first column to the rest of the colors in the palette. </figcaption></p></figure>

5.  Set a threshold value $T$. Then, for each row, compute the error, for each $J=1,..., (L-1),$ as follows,

$$e_k^2(T,P) = \sum_{j=0}^J p(r_k )\Delta^2(r_k,r_j).$$

-	If the error, $e_k^2(T,P)< T$, associated with color $r_k$,  then,
discard colors $j = 1,..., J$ from the Palette $P$, 
-	otherwise keep the colors in the palette. 

6. Adjust $T$ to minimize total error,

$$E(P,T) = \sum_{\forall k} e_k^2(T,P)$$

7. Set the new palette,  $P$ to the colors, which minimizes $E(P,T).$
8. Assign the the pixels with the discarded colors to the most similar color in the palette and generate the quantized image.

The above algorithm generates a quantized image with the most frequent colors and eliminates the colors which are most similar to the most frequent colors.



## 3. Reducing the Coding Redundancy, in the Symbol Encoder

Once we eliminate the interpixel redundancy and psychovisual redundancies in the mapper and quantizer, the last step of compression is to efficiently code the colors. The coding redundancy is minimized in the symbol encoder by coding the colors, which remain in the  palette.


There are two major approaches to code the colors:

**1. Fixed length coding,** where we represent each pixel by m-bits for $2^m$ colors. Thus, we need mNM bits for storage of the image.

**2. Variable length coding,** where we represent frequent colors by less number of bits and  rare colors by more number of bits.

Variable length coding enables us to reach less number of average bits/pixel, defined below;

$$L_{av} = \sum_{k=0}^{L-1} l(r_k)p(r_k),$$
where $l(r_k]$ is the code length of color $r_k$. In this case we need  $MNL_{av}$ bits to store the image.

### Huffman Coding

Huffman coding was developed by [David A. Huffman](https://en.wikipedia.org/wiki/David_A._Huffman),  while he was a student at MIT. His instructor, [Robert Fano](https://en.wikipedia.org/wiki/Robert_Fano) gave the students a choice of a term paper or a final exam. The  term paper was about finding the most efficient binary coding schema. Huffman could not find any efficient code and he was about to give up and start studying for the final exam. Unexpectedly,  he discovered the idea of using a frequency-sorted binary tree. He, also, proved that his method was the most efficient  among the methods, which encode symbols, separately.  He published his pioneering work in a paper,called  ["A Method for the Construction of Minimum-Redundancy Codes"](https://ieeexplore.ieee.org/document/4051119/authors#authors).

Huffman coding is applied in a wide range of applications of computer science and engineering. In the context of image compression, it is used to code the colors after the inter pixel and psychovisual redundancies are eliminated. Once the color palette of an image is obtained, Huffman coding assigns the shortest code length to the most frequent color. As the relative frequency of the colors decrease the corresponding code length of the codeword increases.

Huffman coding has two major passes:
1. Forward pass and
2. Backward pass.

In the forward pass,
1. Probability of occurrence of the colors in the palette is sorted from the most frequent to the least frequent. 
2. Two of the lowest probabilities of the colors are combined by adding them.
3. Step 2 is repeated until only two probabilities remain.

In the backward pass, the codewords are assigned to each reduced source, going backwards. We start by assigning  0 and 1 to the reduced probabilities, at the last step of source reduction. Then as we move backwards, we keep adding 0 and 1 to the unfolded probabilities.

Let's explain the Huffman coding algorithm in a simple example.

Suppose that we have an image of 6 colors, each of which has the following relative frequency of occurrence:

$p(r_0)= 0.1$, $p(r_1)= 0.4$, $p(r_2)= 0.06$ , $p(r_3)= 0.1$ , $p(r_4)= 0.04$, $p(r_5)= 0.3.$

In the forward pass first we sort the probabilities, as shown in Table 1.

In the backward pass, we start from the last column of the source reduction asn assign codes 0 and 1 to the high and low probability values. If we use a fixed length coding, we would allocate $L_{av} =3$ bits per pixel to code 6 colors. On the other hand, Huffman coding requires the average code length as,


$$L_{av} = \sum l(r_k) p(r_j)= 3 \times 0.1 + 0.4 +5 \times 0,6 + 4 \times 0.1 + 5\times 0.04 +2\times 0.3 = 2.2 \text{bits/pixel}$$  

**Note:** Huffman coding assigns only one bit to the most frequent color. As the frequency of occurrence decreases, the code length increases to 5. Average code length per pixel. Thus it is a variable length coding schema. Both coding and decoding can be implemented by a look up table. Decoding can be done unambiguously.

Once we hit a match,  we remove the merged symbols from the sequence and start from the left most symbol again, until we exhaust all the symbols in the string. As a result, we obtain the following sequence:

$$r_2, r_0, r_1, r_1, r_5.$$

# Image Fidelity

When we develop a lossy compression algorithm, there is a compromise between the compression ratio and the image fidelity. As we increase the compression ratio we start to lose the information content, thus the quality of the decompressed image. Therefore, we need to measure the quality of the decompressed images.

There are two types of image fidelity metrics:

**1. Subjective Measures:** We design experiments with a group of people and ask their opinion about how close the generated image to the original one. We may ask people to rate the quality of the decompressed image as follows:

|Value| Rating | Description|
|--|--|--|
| 1 | Excellent | An image of extremely high quality, as good as you could desire|
| 2 | Fine| An image of high quality, providing enjoyable viewing. Interference is not objectionable. |
| 3 | Passable|An image of acceptable quality. Interference is not objectionable. |
| 4 | Marginal|An image of poor quality; you wish you could improve it. Interference is somewhat objectionable. |
| 5 | Inferior|A very poor image, but you could watch it. Objectionable interference is definitely present. |
| 6 | Unusable|An image so bad that you could not watch it. |


We can use statistical methods to evaluate the answers of study groups to develop a subjective measure for image fidelity. For example, we can estimate the mean values and standard deviations of each rating and apply some statistical significance tests to the results. Although the subjective measures provide a measure which is consistent with the human visual system, the measure depends on the evaluations of the study group. Therefore, it is difficult to generate a universal measurement to assess the image fidelity.

**2. Objective Measures**: We can define some  mathematical metrics to measure the image fidelity. Among them the most popular metrics are the followings:

1. Sum of Squared error: Suppose that the original image and decompressed image are $f(x,y)$ and $f_c(x,y)$, respectively. Then the sum of square error is defined as, 

$$SSE (f(x,y) =\sum_x \sum_y [f(x,y) - f_c(x,y)]^2$$

2. Signal to Noise Ratio: This measure is adopted from signal processing to image processing. It measures the ratio between the sum of the pixel brightness values of original image and sum of square error, as given below:

$$SNR (f(x,y) = \frac{\sum_x \sum_y f(x,y)}{\sum_x \sum_y  [f(x,y) - f_c(x,y)]^2}.$$

The objective metrics do not always provide consistent results to the human visual system. However, it removes human subjectivity.





# Chapter Summary

How can we compress an image without loosing its information content? How can we define and measure the information content of an image? 

 In this chapter, we introduced a formal definition for measuring the information content of an image, based on the Information Theory, suggested by Claude Shannon. We defined self information, which measures the information stored in a single pixel. We also defined entropy, which measures the average information contained in an image. Then, by using the coding theorem of Information Theory, we established a lower bound for compressing an image. This theorem basically states that the minimum amount of storage space for an image is bounded by its entropy. However, there are more than one definition of entropy, depending on the definition of the multivariate random variables and its associated joint distribution functions of an image. In the simplest case, we define the first order entropy, using the normalized color histogram of the image.  If we define the joint distribution of the colors between two neighboring pixels, we can define second order entropy and so on. Higher order entropies reduces lower bound for compression. However, they are rather difficult to estimate.

 Once we obtain a lower bound for the required storage space using the entropy,  we can develop algorithms to reduce redundancies by encoding the image. The encoder consists of three major modules. The first one, called the mapper, reduces interpixel redundancy. The second one, called the quantizer, to reduces physchovisual redundancy. The third one is called symbol encoder, reduces coding redundancy.
 At  the output of the encoder we measure the quality of the compressed image using image fidelity metrics, based on subjective and objective criteria.
